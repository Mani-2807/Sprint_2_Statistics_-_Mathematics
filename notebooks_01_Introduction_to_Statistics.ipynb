#  Introduction to Statistics

## 1. What is Statistics?

Statistics is the process of collecting, organizing, analyzing,
interpreting, and presenting data to understand patterns and make
better decisions.

In simple words, statistics helps us convert raw data into useful
information.

For example, if we have the marks of 100 students, statistics can help
us calculate the average marks, identify the highest and lowest marks,
understand how the marks are distributed, and find patterns in the data.

### Example

Suppose the marks of five students are:

60, 70, 75, 80, 90

Instead of looking at all the values separately, statistics allows us
to summarize the data.

The average mark is:

(60 + 70 + 75 + 80 + 90) / 5 = 75

Therefore, the average mark of the students is 75.

This shows how statistics can convert a collection of numbers into
meaningful information.

```python
marks = [60, 70, 75, 80, 90]

average = sum(marks) / len(marks)

print("Marks:", marks)
print("Average marks:", average)
```

### Interpretation

The program calculates the average of the students' marks.

The result is 75, which represents the central value of the given
marks.

### AI/ML Application

Statistics is used in AI/ML to understand datasets before building
models. It helps identify patterns, distributions, unusual values,
relationships between variables, and other characteristics of data.

## 2. Why Statistics is Important in AI/ML

Statistics is important in AI/ML because machine learning models learn
from data.

Before training a model, we need to understand the data. Statistics
helps us summarize the data, identify patterns, detect outliers,
understand relationships between features, and evaluate model results.

Some important uses of statistics in AI/ML are:

- Understanding the dataset  
- Handling missing values  
- Detecting outliers  
- Understanding data distributions  
- Measuring relationships between variables  
- Selecting useful features  
- Scaling and transforming data  
- Evaluating machine learning models  
- Testing whether observed results are meaningful

### Real-World Example

Suppose an online shopping company has customer data containing:

- Age  
- Income  
- Number of previous purchases  
- Time spent on the website  
- Amount spent  

Statistics can help the company understand which variables are related
to customer spending.

An AI/ML model can then use these patterns to predict whether a customer
is likely to purchase a product.

## 3. Statistics vs Machine Learning

Statistics and Machine Learning are closely related, but their main
focus can be different.

Statistics mainly focuses on understanding data, relationships,
uncertainty, and drawing conclusions from data.

Machine Learning focuses on using data to learn patterns and make
predictions or decisions on new data.

### Simple Difference

Statistics → Understand and interpret data.

Machine Learning → Learn patterns from data and make predictions.

### Example

Suppose we have data about house sizes and house prices.

Statistics can help us understand whether house size is related to
house price and how strong that relationship is.

Machine Learning can use house size and other features to learn from
existing houses and predict the price of a new house.

Therefore, statistics provides many concepts that are useful for
understanding and preparing data for machine learning.

## 4. Descriptive Statistics

Descriptive statistics is used to summarize and describe the important
characteristics of a dataset.

It tells us what the data looks like without making conclusions about
a larger population.

Common descriptive statistics include:

- Mean  
- Median  
- Mode  
- Minimum  
- Maximum  
- Range  
- Variance  
- Standard deviation  

```python
import numpy as np

data = [10, 20, 30, 40, 50]

print("Mean:", np.mean(data))
print("Median:", np.median(data))
print("Minimum:", np.min(data))
print("Maximum:", np.max(data))
print("Range:", np.max(data) - np.min(data))
```

### Interpretation

The dataset contains values from 10 to 50.

The mean and median are both 30, which gives us an idea of the center
of the data.

The range is 40, showing the difference between the largest and
smallest values.

### AI/ML Application

Descriptive statistics is commonly used during Exploratory Data
Analysis (EDA) to understand a dataset before training a machine
learning model.

## 5. Inferential Statistics

Inferential statistics is used to make conclusions or predictions about
a larger population based on information obtained from a sample.

Instead of collecting data from every member of a population, we study
a smaller sample and use statistical methods to make conclusions about
the population.

### Example

Suppose a company has 10,000 customers.

Instead of asking all 10,000 customers about their satisfaction, the
company may select 500 customers and analyze their responses.

The results from the sample can be used to estimate the satisfaction
level of the larger customer population.

## 6. Population

A population is the complete group of people, objects, events, or
observations that we are interested in studying.

### Example

If we want to study the salaries of all employees in a company, all
employees in that company form the population.

Population does not necessarily mean people. It can be any complete
group of observations relevant to a study.

## 7. Sample

A sample is a smaller subset selected from a population for analysis.

Studying an entire population can sometimes be expensive, time
consuming, or difficult. Therefore, a sample is selected to represent
the population.

### Example

If a company has 10,000 employees and we select 500 employees to study
their salaries, the 500 employees are the sample.

### Population vs Sample

Population → Complete group being studied.

Sample → Smaller group selected from the population.

### AI/ML Example

A dataset containing all available customer transactions can represent
a population of transactions for a particular study.

A smaller randomly selected group of transactions can be used as a
sample for analysis or experimentation.

## 8. Census

A census means collecting information from every member of the
population being studied.

In other words, instead of selecting a sample, we study the complete
population.

### Example

If a company has 1,000 employees and collects salary information from
all 1,000 employees, it is conducting a census.

### Difference

Census → Data is collected from the entire population.

Sampling → Data is collected from only a portion of the population.

## 9. Sampling

Sampling is the process of selecting a smaller group of observations
from a larger population.

Sampling is useful when collecting information from the entire
population is expensive, time-consuming, or impractical.

### Example

A company has 100,000 customers and wants to understand customer
satisfaction.

Instead of surveying all 100,000 customers, it may select 5,000
customers as a sample.

The selected customers should represent the larger population as well
as possible.

## 10. Variables

A variable is a characteristic or property that can have different
values for different observations.

### Examples

- Age
- Height
- Salary
- Gender
- Temperature
- Number of purchases

For example, if we study five students, their ages may be:

18, 19, 20, 18, 21

Here, age is a variable because its value can differ between students.

## 11. Features

A feature is an input variable used by a machine learning model to
learn patterns and make predictions.

Features are also commonly called input variables or predictors.

### Example

Suppose we want to predict the price of a house.

Possible features are:

- Area
- Number of bedrooms
- Number of bathrooms
- Location
- Age of the house

The machine learning model uses these features to predict the house
price.

### AI/ML Connection

In machine learning, selecting useful features is important because
good features can help a model learn meaningful patterns from data.

## 12. Observation

An observation is a single individual record or data point in a
dataset.

### Example

Consider this student dataset:

| Name | Age | Marks |
|------|-----|-------|
| Ravi | 20 | 85 |
| Anu | 21 | 90 |
| Priya | 20 | 78 |

Each row represents one observation.

Therefore:

- Ravi's record → One observation
- Anu's record → One observation
- Priya's record → One observation

## 13. Dataset

A dataset is a structured collection of observations and variables.

A dataset is often represented as rows and columns.

- Rows → Observations
- Columns → Variables or Features

### Example

| Age | Income | Purchased |
|-----|--------|-----------|
| 22 | 30000 | Yes |
| 35 | 50000 | No |
| 28 | 42000 | Yes |

Here:

- There are 3 observations.
- Age, Income, and Purchased are variables.

## 14. Quantitative Data

Quantitative data represents numerical values that can be measured or
counted.

Examples include:

- Age
- Height
- Weight
- Salary
- Temperature
- Number of products sold

### Example

The ages of five people are:

20, 25, 30, 35, 40

These are quantitative data because they are numerical values.

## 15. Qualitative Data

Qualitative data describes categories, characteristics, or qualities
rather than numerical measurements.

Examples include:

- Gender
- Color
- Product category
- Country
- Customer feedback category

### Example

Customer preferences may be:

Red, Blue, Green

These values describe categories rather than numerical measurements.

## 16. Continuous Data

Continuous data can take any value within a range and is usually
obtained through measurement.

Examples include:

- Height
- Weight
- Temperature
- Distance
- Time

### Example

A person's height could be:

165.2 cm

or

165.25 cm

or

165.253 cm

There can be many possible values between two measurements.

## 17. Discrete Data

Discrete data consists of countable values.

It usually represents quantities that can be counted.

Examples include:

- Number of students
- Number of products
- Number of calls
- Number of website visits

### Example

A shop may receive:

10, 11, 12, or 13 orders.

It cannot receive 10.5 orders.

Therefore, the number of orders is discrete data.

## 18. Structured Data

Structured data is data organized in a predefined format, usually in
rows and columns.

Examples include:

- SQL databases
- Excel spreadsheets
- CSV files
- Tables

### Example

| Customer_ID | Age | Purchase |
|-------------|-----|----------|
| 101 | 25 | 500 |
| 102 | 30 | 750 |
| 103 | 28 | 620 |

The data has clearly defined rows and columns, so it is structured
data.

### AI/ML Application

Structured data is commonly used for machine learning problems such as
classification, regression, and prediction.

## 19. Unstructured Data

Unstructured data does not follow a fixed tabular structure.

Examples include:

- Images
- Videos
- Audio
- Text documents
- Social media posts
- Emails

### Example

A collection of customer product reviews is mostly unstructured text.

An AI system can process this text using Natural Language Processing
(NLP) techniques to identify customer sentiment.

Similarly, images can be processed using computer vision techniques.

## 20. Structured vs Unstructured Data

### Structured Data

- Organized in rows and columns
- Has a predefined format
- Easy to store in relational databases
- Examples: SQL tables, Excel files, CSV files

### Unstructured Data

- Does not follow a fixed tabular format
- Requires specialized techniques for processing
- Examples: Images, videos, audio, and text

### AI/ML Connection

Machine learning can work with both structured and unstructured data.

Structured data is commonly used with traditional machine learning
algorithms, while unstructured data often requires techniques such as
Computer Vision, Natural Language Processing, or Deep Learning.
