#  Measures of Central Tendency

Measures of central tendency are statistical methods used to find a
central or typical value in a dataset.

The main measures covered in this are:

- Mean
- Median
- Mode
- Weighted Mean
- Geometric Mean
- Harmonic Mean

These measures help us summarize a large amount of data using a
representative value

## 1. Mean

The mean is the average value of a dataset.

It is calculated by adding all the values and dividing the sum by the
number of values.

### Formula

Mean = Sum of all values / Number of values

Mathematically:

Mean = (x₁ + x₂ + x₃ + ... + xₙ) / n

where:

- x represents an individual value
- n represents the total number of values

###  Example

Consider the following marks:

60, 70, 80, 90, 100

Mean = (60 + 70 + 80 + 90 + 100) / 5

Mean = 400 / 5

Mean = 80

Therefore, the average mark is 80.

```python
data = [10, 20, 30, 40, 50]

mean = sum(data) / len(data)

print("Mean:", mean)
```

### Code Explanation

- `data` stores the values.
- `sum(data)` calculates the total of all values.
- `len(data)` counts the number of values.
- The total is divided by the number of values.
- The result is the mean.

### Business Example

A company records its monthly sales:

₹10,000, ₹20,000, ₹30,000, ₹40,000, ₹50,000

The company can calculate the mean sales to find its average monthly
sales.

Mean sales = ₹30,000

### AI/ML Example

Mean is commonly used in data preprocessing.

For example, if some values are missing in a numerical feature, the
mean can sometimes be used to replace the missing values.

Mean is also used in statistical calculations and machine learning
evaluation metrics.


### Interpretation

The mean represents the average value of the dataset.

In this example, the mean is 30, which means that the average value of
the five observations is 30.

The mean can be affected by very large or very small values.

## 2. Median

### Definition

Median is the middle value of a dataset after arranging the values in
ascending or descending order.

### Formula

If the number of values is odd:

Median = Middle value

If the number of values is even:

Median = Average of the two middle values

### Example

Consider:

10, 20, 30, 40, 50

The middle value is 30.

Therefore:

Median = 30

```python
data = [10, 20, 30, 40, 50]

data.sort()

n = len(data)

median = data[n // 2]

print("Median:", median)
```

### Code Explanation

- `data` stores the values.
- `sort()` arranges the values in ascending order.
- `len(data)` finds the number of values.
- `n // 2` finds the middle position.
- The value at the middle position is the median.


### Business Example

Suppose the salaries of five employees are:

₹20,000, ₹25,000, ₹30,000, ₹35,000, ₹1,00,000

The very high salary can increase the mean.

The median salary is ₹30,000 and may give a better idea of the typical
salary in this example.

### AI/ML Example

Median can be used to handle missing values in numerical features.

For example, if a dataset contains an income feature with extreme
values, the median can sometimes be used to fill missing income values
because it is less affected by extreme values than the mean.

### Interpretation

The median represents the middle value of an ordered dataset.

It is less affected by extreme values than the mean.

Therefore, the median can be useful for datasets that contain
outliers.

## 3. Mode

### Definition

Mode is the value that occurs most frequently in a dataset.

### Formula

There is no mathematical formula for mode.

Mode = Most frequently occurring value

### Example

Consider:

10, 20, 20, 30, 40

The value 20 occurs two times.

Therefore:

Mode = 20

```python
from collections import Counter

data = [10, 20, 20, 30, 40]

frequency = Counter(data)

mode = frequency.most_common(1)[0][0]

print("Mode:", mode)
```

### Code Explanation

- `Counter(data)` counts how many times each value occurs.
- `most_common(1)` finds the most frequently occurring value.
- `[0][0]` gets the value itself.
- The result is the mode.


### Business Example

A clothing store records the shirt sizes purchased by customers:

M, L, M, S, M, L, XL

The mode is M because M occurs most frequently.

The store can use this information to understand the most commonly
purchased shirt size.

### AI/ML Example

Mode can be used with categorical data.

For example, suppose a dataset contains:

Hyderabad, Mumbai, Hyderabad, Delhi, Hyderabad

The mode is Hyderabad.

Mode can sometimes be used to replace missing values in categorical
features.

### Interpretation

The mode tells us which value appears most frequently.

In this example, 20 occurs more often than the other values, so the
mode is 20.

## 4. Weighted Mean

### Definition

Weighted mean is an average where different values are given
different levels of importance.

### Formula

Weighted Mean = Σ(Value × Weight) / Σ(Weights)

### Example

Suppose:

Marks = 80, Weight = 2

Marks = 90, Weight = 3

Weighted Mean:

= (80 × 2 + 90 × 3) / (2 + 3)

= (160 + 270) / 5

= 430 / 5

= 86

Therefore, the weighted mean is 86.

```python
values = [80, 90]
weights = [2, 3]

weighted_mean = sum(v * w for v, w in zip(values, weights)) / sum(weights)

print("Weighted Mean:", weighted_mean)
```

### Code Explanation

- `values` stores the marks.
- `weights` stores the importance of each mark.
- `zip()` pairs each value with its weight.
- Each value is multiplied by its weight.
- The weighted values are added together.
- The result is divided by the total weight.

### Business Example

A company evaluates an employee using two scores:

Technical Skills = 80, Weight = 3

Communication = 90, Weight = 2

Technical skills have a greater weight, so they have a greater effect
on the final score.

### AI/ML Example

Weighted averages are useful when different observations or values
should contribute differently to a calculation.

For example, different data points may be assigned different weights
when calculating an overall statistical measure.

### Interpretation

The weighted mean gives more importance to values with higher weights.

In this example, the exam has a weight of 3, while the assignment has
a weight of 2. Therefore, the exam has a greater effect on the final
average.

## 5. Geometric Mean

### Definition

Geometric mean is an average calculated by multiplying all the values
and taking the nth root, where n is the number of values.

### Formula

Geometric Mean = (x₁ × x₂ × ... × xₙ)^(1/n)

### Example

Consider:

4, 16

Geometric Mean:

= √(4 × 16)

= √64

= 8

Therefore, the geometric mean is 8.

```python
data = [4, 16]

product = 1

for value in data:
    product = product * value

geometric_mean = product ** (1 / len(data))

print("Geometric Mean:", geometric_mean)
```

### Code Explanation

- `data` stores the values.
- `product` starts with 1.
- The loop multiplies all the values.
- `len(data)` gives the number of values.
- `product ** (1 / len(data))` calculates the required root.
- The result is the geometric mean.

### Business Example

Suppose an investment grows over multiple years.

Because investment growth is based on multiplying growth factors over
time, the geometric mean can be used to calculate an average growth
rate.

### AI/ML Example

Geometric mean can be useful when combining ratios or multiplicative
changes.

For example, it can be used when analyzing performance or growth
across multiple periods.

### Interpretation

The geometric mean is 8 for the values 4 and 16.

It is useful when the values are related through multiplication,
growth, or ratios rather than simple addition.

## 6. Harmonic Mean

### Definition

Harmonic mean is an average that is commonly used for rates and ratios.

### Formula

Harmonic Mean = n / Σ(1/x)

where:

n = Number of values
x = Individual value

### Example

Consider:

60 and 40

Harmonic Mean:

= 2 / ((1/60) + (1/40))

= 48

Therefore, the harmonic mean is 48.

```python
data = [60, 40]

n = len(data)

harmonic_mean = n / sum(1 / value for value in data)

print("Harmonic Mean:", harmonic_mean)
```

### Code Explanation

- `data` stores the values.
- `len(data)` finds the number of values.
- `1 / value` calculates the reciprocal of each value.
- `sum()` adds all the reciprocals.
- `n` is divided by the sum of the reciprocals.
- The result is the harmonic mean.

### Business Example

Suppose a company measures data transfer rates at different times.

When working with rates, the harmonic mean can be useful for finding
an appropriate average rate

### AI/ML Example

The harmonic mean is used in machine learning when combining
Precision and Recall.

The F1-score is the harmonic mean of Precision and Recall.

F1 = 2 × (Precision × Recall) / (Precision + Recall)

### Interpretation

The harmonic mean of 60 and 40 is 48.

The harmonic mean is useful when dealing with rates or ratios.

It is especially useful when a lower value should have a stronger
effect on the final average.
