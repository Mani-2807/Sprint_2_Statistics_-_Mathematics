# correlation & covariance

## 1. Correlation

### Definition

Correlation is a statistical measure that shows the relationship between two variables.

It tells us:
- Whether two variables move together.  
- The direction of their relationship.  
- The strength of their relationship.  

The correlation value ranges from -1 to +1.

- +1 → Perfect positive correlation  
- 0 → No correlation  
- -1 → Perfect negative correlation  

### Formula

For Pearson correlation:

r = Cov(X, Y) / (σX × σY)

Where:  
- r = correlation coefficient  
- Cov(X,Y) = covariance between X and Y  
- σX = standard deviation of X  
- σY = standard deviation of Y  

```python
import numpy as np

study_hours = [1, 2, 3, 4, 5]
marks = [20, 40, 60, 80, 100]

correlation = np.corrcoef(study_hours, marks)[0, 1]

print("Correlation:", correlation)
```

### 🏢 Business Example

A company can study the relationship between **advertising spending and sales**. If sales generally increase when advertising spending increases, the two variables have a positive correlation.

### 🤖 AI/ML Example

In a house price prediction model, **house size and house price** may have a positive correlation. Correlation helps identify relationships between features and the target variable during Exploratory Data Analysis (EDA).

House Size ↑ → House Price ↑

## Interpretation

A correlation value close to +1 means that the two variables have a strong positive relationship.

A value close to -1 means a strong negative relationship.

A value close to 0 means little or no linear relationship.

Correlation shows association between variables, but it does not prove causation.


## 2. Covariance

### Definition

Covariance measures the direction in which two variables change together.

It tells us whether two variables tend to increase or decrease together.

- Positive covariance → Variables move in the same direction.
- Negative covariance → Variables move in opposite directions.
- Covariance near 0 → Little or no linear relationship.

### Formula

Population covariance:

Cov(X,Y) = Σ[(Xi - X̄)(Yi - Ȳ)] / N

Where:
- Xi = individual value of X
- Yi = individual value of Y
- X̄ = mean of X
- Ȳ = mean of Y
- N = number of observations

```python
import numpy as np

X = [1, 2, 3]
Y = [2, 4, 6]

covariance = np.cov(X, Y, ddof=0)[0, 1]

print("Covariance:", covariance)
```

## Business Example

A company can calculate the covariance between advertising expenditure and sales revenue.

Positive covariance indicates that advertising expenditure and sales revenue tend to increase or decrease together.

## AI/ML Example

Covariance is an important concept in machine learning and statistics.

It is used in covariance matrices and is also an important part of Principal Component Analysis (PCA).

## Interpretation

Covariance tells us the direction of the relationship between two variables.

However, its value depends on the scale of the variables.

Correlation is easier to interpret because it always ranges from -1 to +1.

# 3. Positive Correlation

## Definition

Positive correlation occurs when two variables generally move in the same direction.

When one variable increases, the other variable also tends to increase.

The correlation coefficient is between 0 and +1.

## Numerical Example

Study Hours = [1, 2, 3, 4, 5]

Marks = [20, 40, 60, 80, 100]

As study hours increase, marks also increase.

Therefore, the variables have a strong positive correlation.

```python
import numpy as np

study_hours = [1, 2, 3, 4, 5]
marks = [20, 40, 60, 80, 100]

correlation = np.corrcoef(study_hours, marks)[0, 1]

print("Correlation:", correlation)
```

## Business Example

A business may observe that when advertising spending increases, sales also increase.

This indicates a positive correlation between advertising spending and sales

## AI/ML Example

In a house price prediction model, house size and house price may have a positive correlation.

Larger houses may generally have higher prices.

## Interpretation

A positive correlation means that the variables generally move in the same direction.

A correlation close to +1 indicates a strong positive relationship.

# 4. Negative Correlation

## Definition

Negative correlation occurs when two variables generally move in opposite directions.

When one variable increases, the other variable tends to decrease.

The correlation coefficient is between -1 and 0.

## Numerical Example

Temperature = [10, 20, 30, 40, 50]

Hot Chocolate Sales = [50, 40, 30, 20, 10]

As temperature increases, hot chocolate sales decrease.

Therefore, the variables have a strong negative correlation.

```python
import numpy as np

temperature = [10, 20, 30, 40, 50]
sales = [50, 40, 30, 20, 10]

correlation = np.corrcoef(temperature, sales)[0, 1]

print("Correlation:", correlation)
```

## Business Example

A company may observe that when the price of a product increases, the number of customers purchasing the product decreases.

This indicates a negative correlation between price and demand.

## AI/ML Example

In a loan prediction dataset, interest rate and loan demand may have a negative relationship.

As the interest rate increases, loan demand may decrease.

## Interpretation

A negative correlation means that the variables generally move in opposite directions.

A correlation close to -1 indicates a strong negative relationship.

# 5. No Correlation

## Definition

No correlation means there is little or no linear relationship between two variables.

The correlation coefficient is close to 0.

This means changes in one variable do not show a clear linear pattern with changes in the other variable.


## Numerical Example

Consider:

X = [1, 2, 3, 4, 5]

Y = [5, 2, 4, 1, 3]

There is no clear linear pattern between X and Y.

Therefore, the variables have a weak correlation.

```python
import numpy as np

X = [1, 2, 3, 4, 5]
Y = [5, 2, 4, 1, 3]

correlation = np.corrcoef(X, Y)[0, 1]

print("Correlation:", correlation)
```

## Business Example

A business may find that employee shoe size and monthly sales have no meaningful relationship.

Changes in shoe size do not provide useful information about sales.

## AI/ML Example

If two features have a correlation close to zero, there may be no clear linear relationship between them.

This information can be useful during feature analysis and Exploratory Data Analysis (EDA).

## Interpretation

A correlation close to 0 indicates little or no linear relationship.

However, a correlation of 0 does not always mean that there is absolutely no relationship. A nonlinear relationship may still exist.

# 6. Pearson Correlation

## Definition

Pearson correlation measures the strength and direction of a linear relationship between two numerical variables.

The Pearson correlation coefficient ranges from -1 to +1.

## Formula

r = Cov(X,Y) / (σX × σY)

Where:

- r = Pearson correlation coefficient
- Cov(X,Y) = Covariance between X and Y
- σX = Standard deviation of X
- σY = Standard deviation of Y

## Interpretation

+1 → Perfect positive correlation

0 → No linear correlation

-1 → Perfect negative correlation

## Numerical Example

Study Hours = [1, 2, 3, 4, 5]

Marks = [20, 40, 60, 80, 100]

The values increase together in a linear pattern.

Therefore, Pearson correlation will be close to +1.

```python
from scipy.stats import pearsonr

study_hours = [1, 2, 3, 4, 5]
marks = [20, 40, 60, 80, 100]

correlation, p_value = pearsonr(study_hours, marks)

print("Pearson Correlation:", correlation)
print("P-value:", p_value)
```

## Business Example

A company can use Pearson correlation to measure the linear relationship between advertising cost and sales revenue.

A high positive correlation indicates that higher advertising spending is associated with higher sales.

## AI/ML Example

Pearson correlation can be used during Exploratory Data Analysis (EDA) to identify numerical features that have strong linear relationships with the target variable.

## Interpretation

A Pearson correlation of 1.0 indicates a perfect positive linear relationship between study hours and marks.

Pearson correlation is mainly useful for measuring linear relationships.

# 7. Spearman Correlation

## Definition

Spearman correlation measures the strength and direction of a relationship using the ranks of the data.

It is useful when:

- Data is ordinal.
- The relationship is monotonic.
- The data may not follow a normal distribution.
- We want to work with rankings.

## Formula

ρ = 1 - [6Σd² / n(n² - 1)]

Where:

- ρ = Spearman correlation
- d = Difference between ranks
- n = Number of observations

## Numerical Example

Consider:

Study Hours = [1, 2, 3, 4, 5]

Marks = [20, 40, 60, 80, 100]

The ranking of both variables is exactly the same.

Therefore, the Spearman correlation is +1.

```python
from scipy.stats import spearmanr

study_hours = [1, 2, 3, 4, 5]
marks = [20, 40, 60, 80, 100]

correlation, p_value = spearmanr(study_hours, marks)

print("Spearman Correlation:", correlation)
print("P-value:", p_value)
```

## Business Example

A company can compare customer satisfaction rankings with product quality rankings.

Spearman correlation can show whether higher product quality rankings are associated with higher customer satisfaction rankings.

## AI/ML Example

Spearman correlation can be useful when a feature has a monotonic relationship with the target but the relationship is not necessarily linear.

It can also be useful when working with ranked or ordinal data

## Interpretation

Spearman correlation focuses on the order or ranking of values rather than their exact numerical distances.

A value close to +1 indicates a strong positive monotonic relationship.

# 8. Correlation Matrix

## Definition

A correlation matrix is a table that shows the correlation between multiple variables.

Each cell contains the correlation value between two variables.

The values range from -1 to +1.

## Interpretation

- Close to +1 → Strong positive relationship
- Close to -1 → Strong negative relationship
- Close to 0 → Weak or no linear relationship

## Numerical Example

Suppose a dataset contains:

- Study Hours
- Marks
- Sleep Hours

A correlation matrix can show the relationship between every pair of these variables.

```python
import pandas as pd

data = {
    "Study_Hours": [1, 2, 3, 4, 5],
    "Marks": [20, 40, 60, 80, 100],
    "Sleep_Hours": [8, 7, 6, 5, 4]
}

df = pd.DataFrame(data)

correlation_matrix = df.corr()

print(correlation_matrix)
```

## Business Example

A company can create a correlation matrix to study relationships between:

- Sales
- Advertising
- Product Price
- Customer Visits
- Revenue

This helps the business identify which variables are strongly related.

## AI/ML Example

Correlation matrices are commonly used during Exploratory Data Analysis (EDA).

They help identify relationships between features and can also help identify highly correlated features that may cause multicollinearity.

## Interpretation

A correlation matrix provides a quick overview of the relationships between multiple variables at the same time.

It is especially useful when a dataset contains many numerical features.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

data = {
    "Study_Hours": [1, 2, 3, 4, 5],
    "Marks": [20, 40, 60, 80, 100],
    "Sleep_Hours": [8, 7, 6, 5, 4]
}

df = pd.DataFrame(data)

correlation_matrix = df.corr()

sns.heatmap(correlation_matrix, annot=True, fmt=".2f")

plt.title("Correlation Heatmap")
plt.show()
```
