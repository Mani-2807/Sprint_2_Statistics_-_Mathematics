# STATISTICAL DATA VISUALIZATION

# 1. Histogram

## Definition

A histogram shows how numerical data is distributed across different ranges.

It groups values into intervals called bins.

## Purpose

- Understand data distribution
- Identify common value ranges
- Detect skewness
- Identify possible outliers

## When to Use

Use a histogram when you want to understand the distribution of numerical data.

## Advantages

- Easy to understand
- Shows data distribution clearly
- Useful for detecting skewness

## Limitations

- Mainly used for numerical data
- Results can change depending on the number of bins

## Business Example

A company can use a histogram to understand the distribution of customer ages.

## AI/ML Example

A histogram can be used to understand the distribution of a numerical feature before training an ML model.

```python
import matplotlib.pyplot as plt

marks = [45, 50, 55, 60, 62, 65, 68, 70, 72, 75, 78, 80, 85, 90]

plt.hist(marks, bins=5)
plt.xlabel("Marks")
plt.ylabel("Number of Students")
plt.title("Distribution of Marks")
plt.show()
```

## Interpretation

The histogram shows how students' marks are distributed across different ranges.

Higher bars indicate that more students fall within that mark range.

# 2. Bar Chart

## Definition

A bar chart compares values between different categories.

## Purpose

- Compare categories
- Show counts or values
- Identify the highest and lowest categories

## When to Use

Use a bar chart when comparing categorical data.

## Advantages

- Simple to understand
- Easy to compare categories

## Limitations

- Not ideal for showing continuous numerical distributions

## Business Example

A company can compare sales across different products.

## AI/ML Example

A bar chart can show the number of observations belonging to different classes.

```python
import matplotlib.pyplot as plt

products = ["Laptop", "Phone", "Tablet", "Watch"]
sales = [50, 80, 40, 30]

plt.bar(products, sales)
plt.xlabel("Products")
plt.ylabel("Sales")
plt.title("Product Sales")
plt.show()
```

## Interpretation

The bar chart compares sales for different products.

The Phone category has the highest sales.

# 3. Line Chart

## Definition

A line chart connects data points with lines to show changes or trends over time.

## Purpose

- Identify trends
- Show changes over time
- Compare time-based values

## When to Use

Use a line chart when the data has a meaningful order, especially time.

## Advantages

- Clearly shows trends
- Good for time-series data

## Limitations

- Not suitable for unrelated categories

## Business Example

A company can track monthly sales.

## AI/ML Example

A line chart can show model accuracy or loss across training epochs.

```python
import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr", "May"]
sales = [100, 120, 150, 140, 180]

plt.plot(months, sales, marker="o")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.title("Monthly Sales")
plt.show()
```

## Interpretation

The line chart shows how sales changed from January to May.

Overall, sales show an increasing trend.

# 4. Scatter Plot

## Definition

A scatter plot shows the relationship between two numerical variables.

## Purpose

- Identify relationships
- Detect correlation
- Identify clusters
- Detect possible outliers

## When to Use

Use a scatter plot when comparing two numerical variables.

## Advantages

- Clearly shows relationships
- Helps identify correlation
- Can reveal outliers

## Limitations

- Difficult to interpret with extremely large datasets
- Only shows relationships between selected variables

## Business Example

A company can compare advertising spending with sales.

## AI/ML Example

A scatter plot can help understand the relationship between an input feature and a target variable.

```python
import matplotlib.pyplot as plt

study_hours = [1, 2, 3, 4, 5, 6]
marks = [35, 45, 55, 65, 75, 90]

plt.scatter(study_hours, marks)
plt.xlabel("Study Hours")
plt.ylabel("Marks")
plt.title("Study Hours vs Marks")
plt.show()
```

## Interpretation

The points show a positive relationship between study hours and marks.

As study hours increase, marks generally increase.

# 5. Pie Chart

## Definition

A pie chart represents parts of a whole using different sections of a circle.

## Purpose

- Show proportions
- Show percentage distribution
- Compare parts of a whole

## When to Use

Use a pie chart when there are only a few categories and their total represents 100%.

## Advantages

- Easy to understand proportions
- Good for simple percentage comparisons

## Limitations

- Difficult to compare many categories
- Not suitable for large numbers of categories

## Business Example

A company can show its sales percentage by product.

## AI/ML Example

A pie chart can show the percentage of observations belonging to different classes.

```python
import matplotlib.pyplot as plt

categories = ["Product A", "Product B", "Product C", "Product D"]
sales = [40, 30, 20, 10]

plt.pie(sales, labels=categories, autopct="%1.1f%%")
plt.title("Sales Distribution")
plt.show()
```

## Interpretation

Product A contributes the largest percentage of total sales.

Product D contributes the smallest percentage.

# 6. Density Plot

## Definition

A density plot shows the shape of the distribution of numerical data.

It provides a smooth representation of how frequently values occur.

## Purpose

- Understand data distribution
- Identify peaks
- Compare distributions
- Understand skewness

## When to Use

Use a density plot when you want a smooth view of numerical data distribution.

## Advantages

- Smooth representation
- Easy to compare distributions

## Limitations

- Can be harder to understand for beginners
- Shape depends on the smoothing method

## Business Example

A company can visualize the distribution of customer spending.

## AI/ML Example

Density plots can help understand the distribution of ML features.

```python
import seaborn as sns
import matplotlib.pyplot as plt

marks = [45, 50, 55, 60, 62, 65, 68, 70, 72, 75, 78, 80, 85, 90]

sns.kdeplot(marks, fill=True)

plt.xlabel("Marks")
plt.title("Density Plot of Marks")
plt.show()
```

## Interpretation

The density curve shows where marks are concentrated.

The highest part of the curve represents the range where observations are more common.

# 7. Box Plot

## Definition

A box plot summarizes numerical data using:

- Minimum
- First Quartile (Q1)
- Median
- Third Quartile (Q3)
- Maximum

It can also help identify outliers.

## Purpose

- Understand data spread
- Find the median
- Identify outliers
- Compare distributions

## When to Use

Use a box plot when analyzing the spread of numerical data.

## Advantages

- Shows median and quartiles
- Helps detect outliers
- Useful for comparing groups

## Limitations

- Does not show every individual data point

## Business Example

A company can compare employee salaries across departments.

## AI/ML Example

A box plot can help identify outliers in an ML feature before model training.

```python
import matplotlib.pyplot as plt

salary = [25000, 28000, 30000, 32000, 35000, 38000, 40000, 100000]

plt.boxplot(salary)
plt.ylabel("Salary")
plt.title("Salary Distribution")
plt.show()
```

## Interpretation

The box plot shows the distribution of salaries.

The value 100000 may appear as an outlier because it is much higher than most other salaries.

# 8. Violin Plot

## Definition

A violin plot combines a box plot with a density plot.

It shows both the summary statistics and the shape of the data distribution.

## Purpose

- Understand distribution
- Compare groups
- Identify concentration of values

## When to Use

Use a violin plot when comparing the distribution of numerical data across groups.

## Advantages

- Shows distribution shape
- Shows median and spread
- Useful for comparing groups

## Limitations

- More difficult to understand than a simple box plot

## Business Example

A company can compare employee salaries across departments.

## AI/ML Example

A violin plot can compare the distribution of an ML feature across different target classes.

```python
import seaborn as sns
import matplotlib.pyplot as plt

data = {
    "Department": ["IT", "IT", "IT", "HR", "HR", "HR"],
    "Salary": [40000, 45000, 50000, 30000, 35000, 40000]
}

sns.violinplot(x="Department", y="Salary", data=data)

plt.title("Salary Distribution by Department")
plt.show()
```

## Interpretation

The violin plot shows how salaries are distributed in each department.

The width of the violin represents where more values are concentrated.

# 9. Pair Plot

## Definition

A pair plot shows relationships between multiple numerical variables at the same time.

It creates scatter plots between variables and distribution plots along the diagonal.

## Purpose

- Explore relationships between multiple features
- Identify correlations
- Understand distributions
- Find possible patterns

## When to Use

Use a pair plot during Exploratory Data Analysis (EDA).

## Advantages

- Shows many relationships at once
- Useful for initial data exploration

## Limitations

- Can become difficult to read with many features
- Can be computationally expensive for large datasets

## Business Example

A company can explore relationships between sales, advertising, customers, and revenue.

## AI/ML Example

Pair plots can help understand relationships between ML features before model training.

```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd

data = {
    "Study_Hours": [1, 2, 3, 4, 5],
    "Marks": [35, 45, 60, 75, 90],
    "Sleep_Hours": [8, 7, 6, 7, 6]
}

df = pd.DataFrame(data)

sns.pairplot(df)

plt.show()
```

## Interpretation

The pair plot shows the relationships between Study Hours, Marks, and Sleep Hours.

It helps us identify which variables may have strong relationships with each other.

# 10. Heatmap

## Definition

A heatmap represents numerical values using different levels of shading.

It is commonly used to visualize correlation matrices.

## Purpose

- Identify relationships
- Display correlation
- Find patterns quickly

## When to Use

Use a heatmap when you want to visualize relationships between many numerical variables.

## Advantages

- Easy to identify strong relationships
- Good visual summary of large tables

## Limitations

- Requires numerical data for correlation heatmaps
- Too many variables can make it difficult to read

## Business Example

A company can use a heatmap to understand relationships between business metrics.

## AI/ML Example

A correlation heatmap can help identify highly correlated features before ML model training.

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

data = {
    "Study_Hours": [1, 2, 3, 4, 5],
    "Marks": [20, 40, 60, 80, 100],
    "Sleep_Hours": [8, 7, 6, 5, 4]
}

df = pd.DataFrame(data)

correlation = df.corr()

sns.heatmap(correlation, annot=True)

plt.title("Correlation Heatmap")
plt.show()
```

## Interpretation

The heatmap uses different shades to represent correlation values.

Values close to +1 indicate strong positive relationships.

Values close to -1 indicate strong negative relationships.

Values close to 0 indicate weak or no linear relationship.
