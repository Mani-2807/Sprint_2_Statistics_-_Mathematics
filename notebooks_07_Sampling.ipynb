# Sampling

Sampling is the process of selecting a smaller group of observations
from a larger population.

Sampling is useful when studying the complete population is
expensive, difficult, or time-consuming.


## 1. Sampling

### Definition

Sampling is the process of selecting a smaller group, called a
sample, from a larger group, called a population.

Instead of studying the entire population, we study the sample
and use it to understand the population.

### Formula

There is no single formula for sampling.

Sample ⊂ Population

### Numerical Example

Suppose a company has 10,000 customers.

Studying all 10,000 customers may take a lot of time.

The company selects 500 customers for a survey.

Population = 10,000 customers

Sample = 500 customers

```python
import numpy as np

population = np.arange(1, 10001)

sample = np.random.choice(population, size=500, replace=False)

print("Population Size:", len(population))
print("Sample Size:", len(sample))
print("First 10 Sample Values:", sample[:10])
```

### Business Example

A company may survey 500 customers instead of surveying all
10,000 customers to understand customer satisfaction.

### AI/ML Example

Machine learning datasets can contain millions of records.

Sampling can be used to create a smaller dataset for initial
analysis and experimentation.

### Interpretation

Sampling saves time and resources while allowing us to learn
about a larger population.

## 2. Random Sampling

### Definition

Random sampling is a method where every member of the population
has an equal or known chance of being selected.

The selection is made randomly.

### Formula

There is no specific probability formula required for basic
random sampling.

For simple random sampling:

Probability of selection = Sample Size / Population Size

### Numerical Example

Population = 1,000 customers

Sample = 100 customers

Probability of a particular customer being selected:

100 / 1000 = 0.10

Therefore, the probability is 10%.

```python
import numpy as np

population = np.arange(1, 101)

sample = np.random.choice(population, size=10, replace=False)

print("Random Sample:", sample)
```

### Business Example

A company randomly selects customers from its customer database
to conduct a satisfaction survey.

### AI/ML Example

Random sampling can be used to create training and testing
subsets from a dataset.

### Interpretation

Random sampling reduces the chance of deliberately selecting
only certain types of observations.

## 3. Stratified Sampling

### Definition

Stratified sampling divides the population into different groups,
called strata, and then selects samples from each group.

The groups are created based on an important characteristic.

Examples:

- Age group
- Gender
- Location
- Income group
- Department

### Formula

Sample from each stratum can be selected proportionally:

Stratum Sample Size =
(Stratum Population / Total Population) × Total Sample 

### Numerical Example

Suppose a company has:

Department A = 600 employees  
Department B = 300 employees  
Department C = 100 employees  

Total = 1,000 employees

If we need a sample of 100 employees:

Department A:

600 / 1000 × 100 = 60

Department B:

300 / 1000 × 100 = 30

Department C:

100 / 1000 × 100 = 10

```python
departments = {
    "A": 600,
    "B": 300,
    "C": 100
}

total_population = sum(departments.values())
sample_size = 100

for department, population in departments.items():
    sample = (population / total_population) * sample_size
    print(department, "Sample Size:", int(sample))
```

### Business Example

A company wants employee feedback while making sure every
department is represented.

It can select employees from each department proportionally.

### AI/ML Example

Stratified sampling is useful when a dataset contains important
groups that should all be represented.

For example, when splitting an imbalanced classification dataset,
stratification can help preserve class proportions.

### Interpretation

Stratified sampling ensures that important groups within a
population are represented in the sample.

## 4. Cluster Sampling

### Definition

Cluster sampling divides the population into groups called
clusters.

Instead of selecting individuals from every cluster, we randomly
select some clusters and study the members within those clusters.

### Formula

There is no single formula for cluster sampling.

The main idea is:

Population → Clusters → Randomly select clusters → Collect data

### Numerical Example

Suppose a company has offices in:

Hyderabad  
Bangalore  
Chennai  
Mumbai  
Delhi  

Each city represents a cluster.

Instead of selecting employees from every city, the company may
randomly select three cities and survey employees from those
cities.

```python
import numpy as np

clusters = ["Hyderabad", "Bangalore", "Chennai", "Mumbai", "Delhi"]

selected_clusters = np.random.choice(
    clusters,
    size=2,
    replace=False
)

print("Selected Clusters:", selected_clusters)
```

### Business Example

A company with branches across many cities can randomly select
some branches and survey employees from those branches.

### AI/ML Example

Cluster sampling can be useful when data naturally exists in
groups, such as:

- Hospitals
- Schools
- Cities
- Stores
- Branch offices

### Interpretation

Cluster sampling is useful when the population is naturally
divided into groups and studying every individual is difficult.

## 5. Systematic Sampling

### Definition

Systematic sampling selects observations at regular intervals
from an ordered population.

First, a starting point is selected.

Then every kth observation is selected.

### Formula

Sampling Interval:

k = Population Size / Sample Size

### Numerical Example

Population = 1,000 customers

Required sample = 100 customers

k = 1000 / 100

k = 10

Therefore, we can select every 10th customer after choosing a
starting point.

```python
population = list(range(1, 101))

sample_size = 10
interval = len(population) // sample_size

sample = population[interval - 1::interval]

print("Sampling Interval:", interval)
print("Systematic Sample:", sample)
```

### Business Example

A quality-control team may inspect every 10th product coming
from a production line.

### AI/ML Example

Systematic sampling can be used to select records from a large
ordered dataset when a regular sample is sufficient.

### Interpretation

Systematic sampling is simple and efficient because observations
are selected at regular intervals.

## 6. Sampling Bias

### Definition

Sampling bias occurs when the selected sample does not properly
represent the population.

As a result, conclusions from the sample may be misleading.

### Formula

There is no single mathematical formula for sampling bias.

The main goal is:

Representative Sample → Reliable Results
Biased Sample → Potentially Misleading Results

### Numerical Example

Suppose a company wants to know whether all employees are happy
with a new policy.

There are:

500 employees

But the company surveys only 50 managers.

The sample does not represent all employees.

Therefore, the results may be biased.

```python
population = 500
managers_surveyed = 50

percentage = (managers_surveyed / population) * 100

print("Employees in Population:", population)
print("Managers Surveyed:", managers_surveyed)
print("Percentage Surveyed:", percentage, "%")
```

### Business Example

If an online company asks only its most active customers for
feedback, the results may not represent occasional customers.

### AI/ML Example

Sampling bias can cause a machine learning model to perform well
for one group but poorly for another group that was underrepresented
in the training data.

### Interpretation

A biased sample can produce biased conclusions and biased
machine learning models.

Good sampling should represent the population as accurately as
possible.

## 7. Sampling Error

### Definition

Sampling error is the difference between a population value and
the value calculated from a sample.

It occurs because we study only a sample instead of the entire
population.

### Formula

Sampling Error = Sample Statistic - Population Parameter

### Numerical Example

Suppose:

Population Mean = 50

Sample Mean = 48

Sampling Error:

48 - 50 = -2

The absolute sampling error is:

2

```python
population_mean = 50
sample_mean = 48

sampling_error = sample_mean - population_mean

print("Population Mean:", population_mean)
print("Sample Mean:", sample_mean)
print("Sampling Error:", sampling_error)
print("Absolute Sampling Error:", abs(sampling_error))
```

### Business Example

A company estimates the average customer spending using a sample
of customers.

The sample average may be slightly different from the actual
average spending of all customers.

### AI/ML Example

Sampling error can affect estimates calculated from training
samples.

Larger and better-selected samples generally help reduce sampling
error.

### Interpretation

Sampling error does not necessarily mean that the sampling was
biased.

It can occur simply because a sample is being used instead of
the complete population.

## 8. Sample Size

### Definition

Sample size is the number of observations included in a sample.

Choosing an appropriate sample size is important because:

Too small → Results may be unreliable.

Too large → More time and resources may be required.

### Formula

A commonly used sample-size formula for estimating a population
proportion is:

n = Z² × p × (1-p) / E²

Where:

n = required sample size  
Z = Z-score for the confidence level  
p = estimated population proportion  
E = margin of error

### Numerical Example

Suppose:

Confidence level = 95%

Z = 1.96

p = 0.5

Margin of error = 0.05

Then:

n = (1.96² × 0.5 × 0.5) / (0.05²)

n ≈ 384.16

Therefore, we need approximately:

384 observations

```python
Z = 1.96
p = 0.5
E = 0.05

sample_size = (Z ** 2 * p * (1 - p)) / (E ** 2)

print("Required Sample Size:", round(sample_size))
```

### Business Example

A company wants to survey customers about a new product.

Instead of asking every customer, it can calculate an appropriate
sample size based on the required confidence level and margin
of error.

### AI/ML Example

Sample size is important when collecting data for machine learning.

A larger and representative dataset generally provides more
reliable information for training and evaluation.

### Interpretation

Sample size should be large enough to provide reliable results
but practical enough to collect and process.

The required sample size depends on factors such as confidence
level, variability, and acceptable margin of error.
