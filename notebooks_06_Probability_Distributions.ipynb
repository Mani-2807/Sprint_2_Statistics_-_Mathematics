# Probability Distributions

A probability distribution describes how the possible values of a
random variable are distributed along with their probabilities.

Probability distributions help us understand uncertain events and
predict possible outcomes.

## 1. Bernoulli Distribution

### Definition

Bernoulli Distribution is used when an experiment has only two
possible outcomes.

The outcomes are usually represented as:

1 → Success  
0 → Failure

The experiment is performed only once.

### Formula

P(X = x) = p^x × (1-p)^(1-x)

Where:

p = probability of success  
x = 0 or 1

### Where Used

Bernoulli Distribution is used for binary outcomes such as:

- Yes / No  
- Success / Failure  
- True / False  
- Pass / Fail

### Real-World Example

Suppose a customer either purchases a product or does not purchase
it.

Purchase → 1  
No Purchase → 0

If the probability of purchase is 0.6:

P(Purchase) = 0.6

P(No Purchase) = 0.4

### AI/ML Example

Bernoulli Distribution can represent binary classification outcomes.

For example:

Spam → 1  
Not Spam → 0

It can also represent whether a customer churns or does not churn.


```python
import numpy as np
import matplotlib.pyplot as plt

p = 0.6

outcomes = [0, 1]
probabilities = [1 - p, p]

plt.bar(outcomes, probabilities)
plt.xlabel("Outcome")
plt.ylabel("Probability")
plt.title("Bernoulli Distribution")
plt.xticks([0, 1])
plt.show()
```

### Interpretation

Bernoulli Distribution represents a single trial with two possible
outcomes.

The probability of success is p, while the probability of failure
is 1-p

## 2. Binomial Distribution

### Definition

Binomial Distribution is used when a Bernoulli experiment is
repeated a fixed number of times.

Each trial has two possible outcomes:

- Success  
- Failure  

### Formula

P(X = k) = C(n,k) × p^k × (1-p)^(n-k)

Where:

n = number of trials  
k = number of successes  
p = probability of success  

### Where Used

Binomial Distribution is used when we want to count the number
of successes in multiple independent trials

### Real-World Example

Suppose a salesperson contacts 5 customers.

Each customer has a 0.6 probability of making a purchase.

We can use Binomial Distribution to calculate the probability
of getting exactly 3 purchases

### AI/ML Example

Binomial Distribution can be used to model the number of successful
binary outcomes over a fixed number of trials.

For example:

- Number of customers who respond to an advertisement  
- Number of successful predictions  
- Number of defective products in a fixed sample  

```python
from math import comb

n = 5
k = 3
p = 0.6

probability = comb(n, k) * (p ** k) * ((1 - p) ** (n - k))

print("Probability of exactly 3 successes:", round(probability, 4))
```

```python
import numpy as np
import matplotlib.pyplot as plt

n = 10
p = 0.5

x = np.arange(0, n + 1)

probabilities = [
    comb(n, k) * (p ** k) * ((1 - p) ** (n - k))
    for k in x
]

plt.bar(x, probabilities)
plt.xlabel("Number of Successes")
plt.ylabel("Probability")
plt.title("Binomial Distribution")
plt.show()
```

### Interpretation

Binomial Distribution tells us the probability of getting a
specific number of successes from a fixed number of independent
trials.

Bernoulli → One trial

Binomial → Multiple 

## 3. Poisson Distribution

### Definition

Poisson Distribution is used to model the number of times an event
occurs during a fixed interval of time or space.

Examples:

- Number of customers arriving per hour  
- Number of calls received per minute  
- Number of website errors per day  

### Formula

P(X = k) = (e^(-λ) × λ^k) / k!

Where:

λ = average number of events  
k = number of events  
e = mathematical constant  

### Where Used

Poisson Distribution is useful when:

- Events occur independently.  
- We count events in a fixed interval.  
- The average event rate is known.  

### Real-World Example

A customer service center receives an average of 4 calls per hour.

Poisson Distribution can be used to estimate the probability of
receiving a particular number of calls in one hour.

### AI/ML Example

Poisson Distribution can be used for count-based data.

Examples:

- Number of clicks  
- Number of customer complaints  
- Number of system failures  
- Number of transactions  

```python
import math

lam = 4
k = 3

probability = (math.exp(-lam) * (lam ** k)) / math.factorial(k)

print("Probability of exactly 3 events:", round(probability, 4))
```

```python
import numpy as np
import matplotlib.pyplot as plt

lam = 4

x = np.arange(0, 11)

probabilities = [
    (np.exp(-lam) * (lam ** k)) / math.factorial(k)
    for k in x
]

plt.bar(x, probabilities)
plt.xlabel("Number of Events")
plt.ylabel("Probability")
plt.title("Poisson Distribution")
plt.show()
```

### Interpretation

Poisson Distribution describes how often an event occurs within
a fixed interval.

It is especially useful for count-based data.

## 4. Uniform Distribution

### Definition

Uniform Distribution is a probability distribution where all
possible values within a specified range have equal probability.

There are two types:

- Discrete Uniform Distribution  
- Continuous Uniform Distribution

### Formula

For a continuous uniform distribution:

f(x) = 1 / (b - a)

Where:

a = minimum value  
b = maximum value

### Where Used

Uniform Distribution can be used when every value in a particular
range is equally likely.

### Real-World Example

Suppose a random number is selected between 1 and 10 and every
number has an equal chance of being selected.

### AI/ML Example

Uniform random values are commonly used when generating random
data and initializing values in some machine learning processes.

```python
import numpy as np
import matplotlib.pyplot as plt

data = np.random.uniform(0, 10, 1000)

plt.hist(data, bins=20)
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.title("Uniform Distribution")
plt.show()
```

### Interpretation

In a uniform distribution, values within the specified range have
approximately equal likelihood.

The distribution does not have a central peak like a normal
distribution.

## 5. Normal Distribution

### Definition

Normal Distribution is a continuous probability distribution
with a bell-shaped curve.

It is symmetric around its mean.

Many naturally occurring measurements approximately follow a
normal distribution.

### Formula

f(x) = 1 / (σ√(2π)) × e^[-(x-μ)² / (2σ²)]

Where:

μ = mean  
σ = standard deviation  

### Where Used

Normal Distribution is commonly used for:

- Heights
- Test scores
- Measurement errors
- Manufacturing measurements
- Statistical analysis

### Real-World Example

Suppose the heights of people in a population approximately
follow a normal distribution.

Most people will have heights near the average, while very short
and very tall values occur less frequently.

### AI/ML Example

Normal Distribution is important in many statistical and machine
learning techniques.

It is commonly used in:

- Data analysis  
- Feature scaling  
- Statistical modeling  
- Gaussian-based models  
- Error analysis



```python
import numpy as np
import matplotlib.pyplot as plt

data = np.random.normal(loc=50, scale=10, size=1000)

plt.hist(data, bins=30)
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.title("Normal Distribution")
plt.show()
```

### Interpretation

A normal distribution is:

- Bell-shaped
- Symmetric
- Centered around the mean

In a normal distribution:

Mean ≈ Median ≈ Mode

## 6. Exponential Distribution

### Definition

Exponential Distribution is used to model the time between
successive events.

It is commonly used when events occur continuously and
independently at a constant average rate.

### Formula

f(x) = λe^(-λx)

Where:

λ = rate of occurrence  
x = time between events  

### Where Used

Exponential Distribution is commonly used for:

- Waiting time  
- Time between customer arrivals  
- Time between system failures  
- Service times

### Real-World Example

Suppose customers arrive at a bank continuously.

Exponential Distribution can be used to model the waiting time
until the next customer arrives.

### AI/ML Example

Exponential Distribution can be useful for modeling time-based
events such as:

- Time between transactions  
- Time between system failures  
- Customer waiting time  

```python
import numpy as np
import matplotlib.pyplot as plt

data = np.random.exponential(scale=2, size=1000)

plt.hist(data, bins=30)
plt.xlabel("Waiting Time")
plt.ylabel("Frequency")
plt.title("Exponential Distribution")
plt.show()
```

### Interpretation

Exponential Distribution is useful for modeling waiting time
between events.

Unlike the normal distribution, it is not symmetric.

It has a long right tail because larger waiting times are possible
but less common.
