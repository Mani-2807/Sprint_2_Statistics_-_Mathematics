# STATISTICAL TESTS

# 1. T-Test

## Definition

A T-Test is a statistical test used to compare means and determine whether there is a significant difference.

It is commonly used when the sample size is small and the population standard deviation is unknown.

## When to Use

- To compare means.
- When the data is approximately normally distributed.
- When the population standard deviation is unknown.

## When Not to Use

- When the data is strongly non-normal.
- When the data is categorical.

## Formula

t = (Sample Mean - Population Mean) / (Sample Standard Deviation / √n)

## Business Example

A company wants to check whether the average sales after a new marketing campaign are different from the previous average sales.

## AI/ML Example

A T-Test can be used to compare the performance of two models and check whether their difference is statistically significant.

```python
from scipy.stats import ttest_1samp

marks = [72, 75, 78, 70, 80]

t_stat, p_value = ttest_1samp(marks, 75)

print("T-Statistic:", t_stat)
print("P-Value:", p_value)
```

## Interpretation

If the p-value is less than 0.05, we reject the null hypothesis.

If the p-value is greater than or equal to 0.05, we fail to reject the null hypothesis.

In this example, the p-value is 1.0, so there is no significant difference from the mean value of 75.

# 2. Z-Test

## Definition

A Z-Test is used to determine whether a sample mean is significantly different from a population mean.

It is generally used when the population standard deviation is known or when the sample size is large.

## When to Use

- Population standard deviation is known.
- Sample size is large.
- Data is approximately normally distributed.

## When Not to Use

- Population standard deviation is unknown with a small sample.
- Data is strongly non-normal.

## Formula

Z = (Sample Mean - Population Mean) / (Population Standard Deviation / √n)

## Business Example

A company knows that the average delivery time is 30 minutes and the population standard deviation is 5 minutes. It can use a Z-Test to check whether a new delivery system changes the average delivery time.

## AI/ML Example

A Z-Test can be used to statistically compare a model's average performance with a known benchmark.

```python
import math
from scipy.stats import norm

sample_mean = 32
population_mean = 30
population_std = 5
sample_size = 100

z_score = (sample_mean - population_mean) / (
    population_std / math.sqrt(sample_size)
)

p_value = 2 * (1 - norm.cdf(abs(z_score)))

print("Z-Score:", z_score)
print("P-Value:", p_value)
```

## Interpretation

The Z-score tells us how many standard deviations the sample mean is away from the population mean.

If the p-value is less than 0.05, the difference is considered statistically significant.

Here, we use a two-tailed test because we are checking for a difference in either direction.

# 3. Chi-Square Test

## Definition

The Chi-Square Test is used to determine whether there is a significant relationship between categorical variables.

## When to Use

- When working with categorical data.
- When checking whether two categorical variables are related.

## When Not to Use

- For comparing numerical means.
- When the expected frequencies are too small.

## Formula

χ² = Σ ((Observed - Expected)² / Expected)

## Business Example

A company can check whether customer gender is related to product preference.

## AI/ML Example

The Chi-Square Test can help identify whether a categorical feature has a significant relationship with a target variable.

```python
from scipy.stats import chi2_contingency

data = [
    [30, 20],
    [20, 30]
]

chi2, p_value, dof, expected = chi2_contingency(data)

print("Chi-Square:", chi2)
print("P-Value:", p_value)
print("Degrees of Freedom:", dof)
```

## Interpretation

If the p-value is less than 0.05, we reject the null hypothesis.

This means there is a statistically significant relationship between the categorical variables.

If the p-value is greater than or equal to 0.05, we do not have enough evidence to say that the variables are related.

# 4. ANOVA

## Definition

ANOVA stands for Analysis of Variance.

It is used to compare the means of three or more groups.

## When to Use

- When comparing three or more groups.
- When the data is numerical.
- When the groups are independent.

## When Not to Use

- When there are only two groups.
- When the data does not satisfy the assumptions of ANOVA.

## Formula

F = Between-Group Variation / Within-Group Variation

## Business Example

A company can compare the average sales produced by three different marketing campaigns.

## AI/ML Example

ANOVA can be used to determine whether the performance of a model differs significantly across different groups.

```python
from scipy.stats import f_oneway

group_A = [70, 72, 75, 73, 71]
group_B = [80, 82, 85, 83, 81]
group_C = [90, 92, 95, 93, 91]

f_stat, p_value = f_oneway(group_A, group_B, group_C)

print("F-Statistic:", f_stat)
print("P-Value:", p_value)
```

## Interpretation

If the p-value is less than 0.05, at least one group mean is significantly different from the others.

ANOVA tells us that a difference exists, but it does not directly tell us which specific groups are different.

# 5. Mann-Whitney U Test

## Definition

The Mann-Whitney U Test is a non-parametric test used to compare two independent groups.

It is useful when the data does not follow a normal distribution.

## When to Use

- Two independent groups.
- Numerical or ordinal data.
- Data is not normally distributed.

## When Not to Use

- When the groups are paired.
- When there are more than two independent groups.

## Business Example

A company can compare customer satisfaction scores from two different branches when the scores are not normally distributed.

## AI/ML Example

It can be used to compare model performance between two independent groups when the data is not normally distributed.

```python
from scipy.stats import mannwhitneyu

group_A = [10, 12, 15, 18, 20]
group_B = [14, 16, 19, 21, 25]

u_stat, p_value = mannwhitneyu(group_A, group_B)

print("U-Statistic:", u_stat)
print("P-Value:", p_value)
```

## Interpretation

If the p-value is less than 0.05, there is a statistically significant difference between the two groups.

The Mann-Whitney U Test does not compare means directly. It compares the distributions/ranks of the two groups.

# 6. Shapiro-Wilk Test

## Definition

The Shapiro-Wilk Test is used to check whether a dataset is approximately normally distributed.

## When to Use

It is useful for checking the normality assumption before applying statistical tests such as a T-Test or ANOVA.

## When Not to Use

It should not be used as the only method for deciding normality, especially with very large datasets.

## Hypotheses

H₀: The data follows a normal distribution.

H₁: The data does not follow a normal distribution.

## Business Example

A company can check whether customer waiting times are approximately normally distributed before selecting an appropriate statistical test.

## AI/ML Example

An ML engineer can use the Shapiro-Wilk Test to check whether a numerical feature approximately follows a normal distribution.

```python
from scipy.stats import shapiro

data = [10, 12, 11, 13, 12, 14, 15, 11, 13, 12]

stat, p_value = shapiro(data)

print("Shapiro-Wilk Statistic:", stat)
print("P-Value:", p_value)
```

## Interpretation

If the p-value is greater than 0.05:

We fail to reject the null hypothesis, so the data does not show strong evidence against normality.

If the p-value is less than 0.05:

We reject the null hypothesis, suggesting that the data is not normally distributed.

The Shapiro-Wilk Test is mainly used to check the normality assumption.
