# Probability

Probability is a branch of mathematics used to measure how likely
an event is to happen.

Probability is very important in AI/ML because many real-world
problems involve uncertainty and predictions.

## 1. Introduction to Probability

### Definition

Probability measures how likely an event is to happen.

The value of probability is always between 0 and 1.

0 → Event is impossible

1 → Event is certain

A probability closer to 1 means the event is more likely to happen.

### Formula

Probability(Event) =
Number of favorable outcomes / Total number of possible outcomes

### Numerical Example

A fair coin has two possible outcomes:

Heads
Tails

Probability of getting Heads:

P(Heads) = 1 / 2

P(Heads) = 0.5

Therefore, there is a 50% chance of getting Heads.


```python
favorable_outcomes = 1
total_outcomes = 2

probability = favorable_outcomes / total_outcomes

print("Probability of Heads:", probability)
```

### Business Example

A company may calculate the probability that a customer will
purchase a product.

### AI/ML Example

A classification model can predict the probability that an email
is spam.

For example:

Spam probability = 0.90

This means the model estimates a 90% probability that the email
is spam.

### Interpretation

Probability helps us measure uncertainty and make decisions
based on how likely an event is to occur

## 2. Events

### Definition

An event is a specific outcome or group of outcomes that we are
interested in.

For example, when rolling a dice:

Getting a 6 is an event.

Getting an even number is also an event.

### Formula

Probability of an event:

P(A) = Favorable Outcomes / Total 

### Numerical Example

A standard dice has six outcomes:

1, 2, 3, 4, 5, 6

Suppose Event A is getting an even number.

Favorable outcomes:

2, 4, 6

Number of favorable outcomes = 3

Total outcomes = 6

P(A) = 3 / 6

P(A) = 0.5

```python
favorable_outcomes = 3
total_outcomes = 6

probability = favorable_outcomes / total_outcomes

print("Probability of getting an even number:", probability)
```

### Business Example

A company may define an event as:

"Customer purchases a product."

The probability of this event can help the company understand
customer behavior.

### AI/ML Example

In classification, predicting "customer will leave" can be
treated as an event.

### Interpretation

An event represents something that may happen and whose
probability we want to calculate

## 4. Independent Events

### Definition

Two events are independent when the occurrence of one event does
not affect the probability of the other event.

### Formula

P(A and B) = P(A) × P(B)

### Numerical Example

Suppose we toss a coin twice.

Probability of getting Heads on the first toss:

P(A) = 1/2

Probability of getting Heads on the second toss:

P(B) = 1/2

Therefore:

P(A and B) = 1/2 × 1/2

P(A and B) = 1/4

= 0.25

```python
probability_heads_first = 0.5
probability_heads_second = 0.5

probability_both_heads = (
    probability_heads_first * probability_heads_second
)

print("Probability of getting Heads twice:", probability_both_heads)
```

### Business Example

A customer purchasing one product may not affect another
independent customer's purchase.

### AI/ML Example

Naive Bayes assumes conditional independence between features
when calculating probabilities.

### Interpretation

Independent events do not influence each other.

Therefore, their joint probability can be calculated by
multiplying their individual probabilities

## 5. Dependent Events

### Definition

Two events are dependent when the occurrence of one event affects
the probability of another event.

### Formula

P(A and B) = P(A) × P(B | A)

Where:

P(B | A) = Probability of B given that A has already occurred.

### Numerical Example

A bag contains:

3 red balls
2 blue balls

Total = 5 balls

Suppose we select two balls without replacement.

Probability of selecting a red ball first:

P(A) = 3/5

After selecting one red ball:

Remaining red balls = 2
Remaining total balls = 4

P(B | A) = 2/4

Therefore:

P(A and B) = 3/5 × 2/4

P(A and B) = 0.30

```python
probability_first_red = 3 / 5
probability_second_red = 2 / 4

probability_both_red = probability_first_red * probability_second_red

print("Probability of selecting two red balls:", probability_both_red)
```

### Business Example

If a customer has already purchased one product, their probability
of purchasing another related product may change.

### AI/ML Example

Dependent variables or features can influence each other in
real-world datasets.

Understanding these relationships is important during feature
analysis.

### Interpretation

In dependent events, the first event changes the probability of
the second event.

## 6. Conditional Probability

### Definition

Conditional probability is the probability of an event occurring
given that another event has already occurred.

### Formula

P(A | B) = P(A and B) / P(B)

Where:

P(A | B) = Probability of A given B
P(A and B) = Probability of both A and B
P(B) = Probability of B

### Numerical Example

Suppose:

P(A and B) = 0.20

P(B) = 0.50

Then:

P(A | B) = 0.20 / 0.50

P(A | B) = 0.40

Therefore, the probability of A given B is 40%.

```python
probability_a_and_b = 0.20
probability_b = 0.50

conditional_probability = probability_a_and_b / probability_b

print("P(A | B):", conditional_probability)
```

### Business Example

An online store may calculate:

Probability that a customer buys a product given that the customer
has viewed the product.

### AI/ML Example

Conditional probability is fundamental to classification models
and especially important in Naive Bayes.

### Interpretation

Conditional probability allows us to update the probability of an
event when we have additional information.

## 7. Bayes' Theorem

### Definition

Bayes' Theorem is used to calculate the probability of an event
when we have new information or evidence.

It helps us update an existing probability using new evidence.

### Formula

P(A | B) = [P(B | A) × P(A)] / P(B)

Where:

P(A | B) = Probability of A given B  
P(B | A) = Probability of B given A  
P(A) = Probability of A  
P(B) = Probability of B  


### Numerical Example

Suppose:

P(A) = 0.20
P(B | A) = 0.80
P(B) = 0.40

Using Bayes' Theorem:

P(A | B) = (0.80 × 0.20) / 0.40

P(A | B) = 0.40

Therefore, the probability of A given B is 40%.

```python
probability_a = 0.20
probability_b_given_a = 0.80
probability_b = 0.40

probability_a_given_b = (
    probability_b_given_a * probability_a
) / probability_b

print("P(A | B):", probability_a_given_b)
```

### Business Example

A company can use Bayes' Theorem to estimate the probability that
a customer will purchase a product after receiving an advertisement.

### AI/ML Example

Bayes' Theorem is the foundation of Naive Bayes classification.

For example, it can be used to calculate the probability that an
email is spam based on the words present in the email.

### Interpretation

Bayes' Theorem helps us update the probability of an event when
new evidence becomes available.

## 8. Probability Rules

### Definition

Probability rules are basic mathematical rules used to calculate
and combine probabilities.

Some important rules are:

1. Complement Rule  
2. Addition Rule  
3. Multiplication Rule  

### Formulas

Complement Rule:

P(A') = 1 - P(A)

Addition Rule:

P(A or B) = P(A) + P(B) - P(A and B)

For independent events:

P(A and B) = P(A) × P(B)

### Numerical Example

Suppose:

P(A) = 0.7

Using the Complement Rule:

P(A') = 1 - 0.7

P(A') = 0.3

Therefore, the probability that A does not occur is 30%.

```python
probability_a = 0.7

probability_not_a = 1 - probability_a

print("P(A):", probability_a)
print("P(Not A):", probability_not_a)
```

```python
print("P(A):", probability_a)
print("P(Not A):", round(probability_not_a, 2))
```

### Business Example

A company may calculate the probability that a customer will
purchase a product and the probability that the customer will
not purchase it.

### AI/ML Example

Probability rules are used throughout machine learning for
classification, prediction, and uncertainty estimation.

### Interpretation

Probability rules help us combine and calculate probabilities
correctly.

The complement rule is especially useful for finding the
probability that an event does not happen.

## 9. Random Variables

### Definition

A random variable is a variable whose value depends on the outcome
of a random experiment.

There are two main types:

1. Discrete Random Variable
2. Continuous Random Variable

A discrete random variable has countable values.

Example:

Number of customers arriving at a store.

A continuous random variable can take values within a range.

Example:

Height of a person.

### Formula

There is no single formula for a random variable.

A random variable maps outcomes of a random experiment to numerical
values.

### Numerical Example

Suppose we toss a coin twice.

Possible outcomes:

HH  
HT  
TH  
TT  

Let X represent the number of Heads.

Then:

HH → X = 2  
HT → X = 1  
TH → X = 1  
TT → X = 0  

Therefore, X can take the values:

0, 1, 2

```python
outcomes = ["HH", "HT", "TH", "TT"]

for outcome in outcomes:
    heads = outcome.count("H")
    print(outcome, "-> Number of Heads:", heads)
```

### Business Example

The number of customers visiting a store each day can be treated
as a random variable.

For example:

X = Number of customers per day

X may be 100, 120, 150, 200, etc.

### AI/ML Example

Random variables are used to represent uncertain values in
probability models and machine learning algorithms.

Examples include:

- Number of purchases
- Customer age
- Temperature
- Model prediction probability

### Interpretation

A random variable converts the outcome of a random experiment
into a numerical value that can be analyzed statistically.

## 10. Expected Value

### Definition

Expected value is the long-run average value we expect from a
random experiment when it is repeated many times.

It is a probability-weighted average of all possible outcomes.

### Formula

E(X) = Σ [x × P(X = x)]

Where:

x = possible outcome

P(X = x) = probability of that outcome

### Numerical Example

Suppose a game has two possible outcomes:

Win → ₹100 with probability 0.5

Lose → ₹0 with probability 0.5

Expected Value:

E(X) = (100 × 0.5) + (0 × 0.5)

E(X) = 50

Therefore, the expected value is ₹50.

```python
outcomes = [100, 0]
probabilities = [0.5, 0.5]

expected_value = 0

for outcome, probability in zip(outcomes, probabilities):
    expected_value += outcome * probability

print("Expected Value:", expected_value)
```

### Business Example

A company can calculate the expected revenue from a new product
by considering different possible sales outcomes and their
probabilities.

### AI/ML Example

Expected value is used in probability-based decision making and
can help estimate the average outcome of uncertain predictions.

For example, expected loss can be used when evaluating the
possible outcomes of a machine learning model.

### Interpretation

Expected value represents the average outcome we would expect
over many repetitions of the same random experiment.

It does not mean that the exact expected value must occur in
every individual experiment.
