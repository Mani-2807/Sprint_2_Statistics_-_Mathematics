## 1. Normal Distribution

### Definition

A Normal Distribution is a continuous probability distribution where most of the data values are concentrated around the center (mean), and fewer values occur as we move away from the center.

It has a bell-shaped curve and is symmetric around the mean.

In a normal distribution:

- Mean = Median = Mode
- Most observations are near the mean.
- Very few observations occur far from the mean.
- The left and right sides of the distribution have the same shape.

### Numerical Example

Suppose the heights of students are approximately:

150, 155, 158, 160, 161, 162, 164, 166, 170

Most values are concentrated around the center, while very small and very large values occur less frequently.

This type of pattern can be represented approximately by a normal distribution.

### Business Example

A company may analyze the delivery times of thousands of orders.

If most deliveries take around 30 minutes and fewer deliveries take much less or much more time, the delivery times may approximately follow a normal distribution.

### AI/ML Example

Normal distributions are commonly used in statistical analysis and machine learning.

For example, a feature such as measurement error may approximately follow a normal distribution. Understanding its distribution helps in preprocessing, analysis, and model assumptions.

### Interpretation

A normal distribution tells us that values are concentrated around the average and become less common as we move farther from the average.

```python
import numpy as np
import matplotlib.pyplot as plt

# Generate normally distributed data
data = np.random.normal(loc=50, scale=10, size=1000)

# Plot histogram
plt.hist(data, bins=30, density=True)
plt.xlabel("Value")
plt.ylabel("Density")
plt.title("Normal Distribution")
plt.show()
```

### Explanation

The program generates 1000 random values with:

- Mean = 50
- Standard deviation = 10

The histogram shows how the values are distributed around the mean.

### Interpretation

The graph has a bell-shaped pattern.

Most values are close to 50, while fewer values are found near the lower and upper ends.

This demonstrates the basic shape of a normal distribution.

## 2. Uniform Distribution

### Definition

A Uniform Distribution is a distribution where all values within a specified range have approximately equal chances of occurring.

Unlike a normal distribution, there is no central value where observations are strongly concentrated.

### Numerical Example

Suppose a random number generator produces numbers between 1 and 10.

Each number has approximately the same probability of being selected.

### Business Example

A company randomly assigns customers to one of several equally sized experimental groups.

If every group has the same probability of receiving a customer, the assignment can be represented using a uniform distribution.

### AI/ML Example

Uniform distributions can be used when generating random values for simulations or initializing certain model parameters.

### Interpretation

A uniform distribution means that values within the defined range have equal probability.

```python
import numpy as np
import matplotlib.pyplot as plt

# Generate uniformly distributed data
data = np.random.uniform(0, 10, 1000)

# Plot histogram
plt.hist(data, bins=20, density=True)
plt.xlabel("Value")
plt.ylabel("Density")
plt.title("Uniform Distribution")
plt.show()
```

### Interpretation

The histogram is approximately flat across the range from 0 to 10.

This means that values throughout the range have approximately equal probability.

## 3. Gaussian Distribution

### Definition

A Gaussian Distribution is another name for the Normal Distribution.

It is a continuous probability distribution characterized by its bell-shaped and symmetric curve.

The shape of the Gaussian distribution is determined mainly by:

- Mean – determines the center of the distribution.
- Standard deviation – determines how spread out the data is.

### Numerical Example

Suppose:

Mean = 100
Standard Deviation = 15

Most values will occur around 100, while values farther away from 100 become less common.

### Business Example

A company may analyze employee performance scores. If most scores are near the average and fewer employees have extremely low or high scores, the scores may approximately follow a Gaussian distribution.

### AI/ML Example

Gaussian distributions are important in many statistical and machine learning techniques, including Gaussian Naive Bayes and Gaussian mixture models.

### Interpretation

Gaussian Distribution and Normal Distribution refer to the same basic distribution.

The mean controls the center, while the standard deviation controls the spread.

```python
import numpy as np
import matplotlib.pyplot as plt

mean = 100
std = 15

data = np.random.normal(mean, std, 1000)

plt.hist(data, bins=30, density=True)
plt.xlabel("Value")
plt.ylabel("Density")
plt.title("Gaussian Distribution")
plt.show()
```

## 4. Left Skew

### Definition

A Left-Skewed Distribution is a distribution where the long tail extends toward the left side.

Most of the observations are concentrated toward the higher values, while a smaller number of low values create the long left tail.

In a typical left-skewed distribution:

Mean < Median

### Numerical Example

Consider exam scores where most students score highly, but a few students score very low.

Example:

20, 70, 80, 82, 85, 87, 90, 92, 95

The few low scores create a tail toward the left.

### Business Example

If most employees have high performance scores but a few employees have very low scores, the distribution may be left-skewed.

### AI/ML Example

Understanding skewness helps identify whether a feature has an asymmetric distribution and whether transformations may be useful during preprocessing.

### Interpretation

A left-skewed distribution has most of its data on the higher-value side and a longer tail toward lower values.

```python
import numpy as np
import matplotlib.pyplot as plt

data = np.random.beta(5, 2, 1000) * 100

plt.hist(data, bins=30)
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.title("Left-Skewed Distribution")
plt.show()
```

## 5. Right Skew

### Definition

A Right-Skewed Distribution is a distribution where the long tail extends toward the right side.

Most observations are concentrated toward the lower values, while a smaller number of high values create the long right tail.

In a typical right-skewed distribution:

Mean > Median

### Numerical Example

Consider salaries in a company.

Most employees may earn moderate salaries, while a small number of senior executives earn very high salaries.

Those very high salaries create a long right tail.

### Business Example

Income, house prices, and customer spending are examples of data that can often be right-skewed.

### AI/ML Example

A highly right-skewed feature may affect some machine learning algorithms. Transformations such as logarithmic transformation can sometimes be used to reduce skewness.

### Interpretation

A right-skewed distribution has most of its observations toward the lower-value side and a long tail toward higher values.

```python
import numpy as np
import matplotlib.pyplot as plt

data = np.random.exponential(scale=2, size=1000)

plt.hist(data, bins=30)
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.title("Right-Skewed Distribution")
plt.show()
```

## 6. Symmetric Distribution

### Definition

A Symmetric Distribution is a distribution where the left and right sides have approximately the same shape around the center.

If we draw a vertical line through the center, both sides would look similar.

For a perfectly symmetric normal distribution:

Mean = Median = Mode

### Numerical Example

Consider:

2, 3, 4, 5, 6, 7, 8

The values are balanced around the center value 5.

### Business Example

If customer satisfaction scores are distributed similarly on both sides of the average score, the distribution may be approximately symmetric.

### AI/ML Example

Checking whether a feature is symmetric helps us understand its distribution before applying statistical techniques or transformations.

### Interpretation

A symmetric distribution has a balanced shape around its center.

```python
import numpy as np
import matplotlib.pyplot as plt

data = np.random.normal(50, 10, 1000)

plt.hist(data, bins=30)
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.title("Symmetric Distribution")
plt.show()
```

## 7. Kurtosis

### Definition

Kurtosis is a statistical measure that describes the shape of the tails of a distribution and how much extreme values may occur compared with a normal distribution.

It helps us understand whether a distribution has relatively heavy or light tails.

There are three common descriptions:

- Mesokurtic – similar tail behavior to a normal distribution.
- Leptokurtic – heavier tails and more extreme values.
- Platykurtic – lighter tails and fewer extreme values.

### Numerical Example

Suppose two datasets have similar means and standard deviations.

If one dataset contains more extreme values than the other, it may have heavier tails and therefore higher kurtosis.

### Business Example

In finance, the returns of an asset may contain occasional extreme gains or losses. Kurtosis can help identify the presence of heavy tails.

### AI/ML Example

Kurtosis can help identify whether a feature contains extreme observations that may require further investigation.

### Interpretation

Kurtosis mainly helps us understand the tail behavior of a distribution and the presence of extreme values.

```python
import numpy as np
from scipy.stats import kurtosis

data = np.random.normal(50, 10, 1000)

kurtosis_value = kurtosis(data)

print("Kurtosis:", kurtosis_value)
```

### Interpretation

The calculated value represents the excess kurtosis of the generated data.

For a normal distribution, excess kurtosis is approximately 0.

Because the data is randomly generated, the exact value may vary slightly each time the program runs.

## 8. Empirical Rule (68–95–99.7)

### Definition

The Empirical Rule describes how data is distributed around the mean in an approximately normal distribution.

For a normal distribution:

- Approximately 68% of data falls within 1 standard deviation of the mean.
- Approximately 95% falls within 2 standard deviations.
- Approximately 99.7% falls within 3 standard deviations.

This is also called the 68–95–99.7 rule.

### Numerical Example

Suppose:

Mean = 100
Standard Deviation = 10

Within 1 standard deviation:

90 to 110 → approximately 68% of observations

Within 2 standard deviations:

80 to 120 → approximately 95% of observations

Within 3 standard deviations:

70 to 130 → approximately 99.7% of observations

### Business Example

A company can use this rule to understand how employee performance measurements are distributed around the average when the data is approximately normal.

### AI/ML Example

The Empirical Rule can help identify unusually distant observations and understand how much data lies close to the mean.

### Interpretation

The farther we move from the mean, the smaller the proportion of observations becomes.

The rule applies specifically to approximately normal distributions.

```python
import numpy as np

mean = 100
std = 10

data = np.random.normal(mean, std, 100000)

within_1 = np.mean((data >= mean - std) & (data <= mean + std))
within_2 = np.mean((data >= mean - 2*std) & (data <= mean + 2*std))
within_3 = np.mean((data >= mean - 3*std) & (data <= mean + 3*std))

print("Within 1 standard deviation:", within_1 * 100, "%")
print("Within 2 standard deviations:", within_2 * 100, "%")
print("Within 3 standard deviations:", within_3 * 100, "%")
```

### Interpretation

The output should be close to:

- 68% within 1 standard deviation
- 95% within 2 standard deviations
- 99.7% within 3 standard deviations

Small differences occur because the dataset is randomly generated.

This demonstrates the Empirical Rule using simulated data.
