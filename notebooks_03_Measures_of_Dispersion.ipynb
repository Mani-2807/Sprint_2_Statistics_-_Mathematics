# Measures of Dispersion

Measures of dispersion are statistical methods used to understand
how spread out or scattered the values in a dataset are.

In this, we will learn:

- Range
- Variance
- Standard Deviation
- Quartiles
- Percentiles
- Interquartile Range (IQR)
- Coefficient of Variation

## 1. Range

### Definition

Range is the difference between the largest value and the smallest
value in a dataset.

It gives a simple idea about how widely the data is spread.

### Formula

Range = Maximum Value - Minimum Value

### Numerical Example

Consider the dataset:

10, 20, 30, 40, 50

Maximum value = 50
Minimum value = 10

Range = 50 - 10

Range = 40

```python
data = [10, 20, 30, 40, 50]

maximum = max(data)
minimum = min(data)

range_value = maximum - minimum

print("Data:", data)
print("Maximum:", maximum)
print("Minimum:", minimum)
print("Range:", range_value)
```

### Business Example

A company records monthly sales:

₹10,000, ₹20,000, ₹30,000, ₹40,000, ₹50,000

The range is ₹40,000.

This shows the difference between the highest and lowest monthly sales.

### AI/ML Example

Range can be used during exploratory data analysis (EDA) to quickly
understand the spread of a numerical feature.

For example, it can show the difference between the minimum and
maximum house prices.

### Interpretation

A larger range indicates a wider spread between the minimum and
maximum values.

A smaller range indicates that the values are closer together.

However, range depends only on two values, so it can be strongly
affected by extreme values.

## 2. Variance

### Definition

Variance measures how far the values in a dataset are spread from
the mean.

A small variance means the values are close to the mean.

A large variance means the values are more spread out.

### Formula

Population Variance:

σ² = Σ(x - μ)² / N

Sample Variance:

s² = Σ(x - x̄)² / (n - 1)

Where:

x = individual value
μ = population mean
x̄ = sample mean
N = population size
n = sample size

### Numerical Example

Consider:

10, 20, 30

Mean:

(10 + 20 + 30) / 3 = 20

Squared differences:

(10 - 20)² = 100
(20 - 20)² = 0
(30 - 20)² = 100

Population Variance:

(100 + 0 + 100) / 3

= 66.67

```python
data = [10, 20, 30]

mean = sum(data) / len(data)

squared_differences = []

for value in data:
    squared_differences.append((value - mean) ** 2)

variance = sum(squared_differences) / len(data)

print("Data:", data)
print("Mean:", mean)
print("Variance:", round(variance, 2))
```

### Business Example

A company compares the variation in monthly sales.

If sales have high variance, monthly sales are changing considerably.

If sales have low variance, monthly sales are relatively stable.

### AI/ML Example

Variance is useful during feature analysis.

For example, a feature with almost no variance contains very little
variation and may provide limited information to a machine learning
model.

### Interpretation

Variance tells us how much the data varies around its mean.

Higher variance → more spread

Lower variance → less spread

## 3. Standard Deviation

### Definition

Standard deviation measures the amount of variation or spread in a
dataset.

It is the square root of variance.

### Formula

Standard Deviation = √Variance

For population data:

σ = √[Σ(x - μ)² / N]

For sample data:

s = √[Σ(x - x̄)² / (n - 1)]

### Numerical Example

Consider:

10, 20, 30

Mean = 20

Population Variance = 66.67

Therefore:

Standard Deviation = √66.67

Standard Deviation ≈ 8.16

```python
import math

data = [10, 20, 30]

mean = sum(data) / len(data)

squared_differences = []

for value in data:
    squared_differences.append((value - mean) ** 2)

variance = sum(squared_differences) / len(data)

standard_deviation = math.sqrt(variance)

print("Data:", data)
print("Variance:", round(variance, 2))
print("Standard Deviation:", round(standard_deviation, 2))
```

### Business Example

A company can use standard deviation to understand how much monthly
sales vary from the average sales.

### AI/ML Example

Standard deviation is important in feature scaling.

For example, StandardScaler uses the mean and standard deviation
to standardize numerical features.

### Interpretation

A small standard deviation means the values are close to the mean.

A large standard deviation means the values are widely spread.

Unlike variance, standard deviation is expressed in the same units
as the original data.

## 4. Quartiles

### Definition

Quartiles divide an ordered dataset into four parts.

The three important quartiles are:

Q1 → First Quartile → 25th percentile

Q2 → Second Quartile → 50th percentile (Median)

Q3 → Third Quartile → 75th percentile

### Formula

There is no single simple arithmetic formula for all quartile
calculation methods. Quartiles are calculated from the ordered
positions of the data.

### Numerical Example

Consider:

10, 20, 30, 40, 50, 60, 70

Q2 (Median) = 40

The lower part is:

10, 20, 30

Q1 = 20

The upper part is:

50, 60, 70

Q3 = 60

Therefore:

Q1 = 20  
Q2 = 40  
Q3 = 60  

```python
import numpy as np

data = [10, 20, 30, 40, 50, 60, 70]

q1 = np.percentile(data, 25)
q2 = np.percentile(data, 50)
q3 = np.percentile(data, 75)

print("Data:", data)
print("Q1:", q1)
print("Q2:", q2)
print("Q3:", q3)
```

### Business Example

A company can use quartiles to understand employee salaries.

For example:

Q1 → salary level below which 25% of employees fall

Q2 → median salary

Q3 → salary level below which 75% of employees fall

### AI/ML Example

Quartiles are useful during data analysis and outlier detection.

They are also used to calculate the Interquartile Range (IQR).

### Interpretation

Quartiles help us understand how the data is distributed across
different sections.

Q1 represents the lower part of the data.

Q2 represents the middle.

Q3 represents the upper part.

## 5. Percentiles

### Definition

A percentile tells us the value below which a certain percentage
of observations fall.

For example:

25th percentile → 25% of values are below this value

50th percentile → 50% of values are below this value

90th percentile → 90% of values are below this value

### Formula

Percentiles are calculated using the position of values in an
ordered dataset.

A common position formula is:

Position = (P / 100) × (n + 1)

Different statistical methods may use different percentile
calculation methods.

### Numerical Example

Suppose student scores are:

50, 60, 70, 80, 90

The 50th percentile represents the middle of the dataset.

The 50th percentile is 70.

This means approximately 50% of the observations are at or below
this value.

```python
import numpy as np

scores = [50, 60, 70, 80, 90]

p50 = np.percentile(scores, 50)
p90 = np.percentile(scores, 90)

print("Scores:", scores)
print("50th Percentile:", p50)
print("90th Percentile:", p90)
```

### Business Example

An online store can use the 90th percentile of delivery time to
understand a high-end delivery time experienced by customers.

### AI/ML Example

Percentiles can be used for feature analysis, threshold selection,
and identifying extreme values.

### Interpretation

A percentile describes the relative position of a value within a
dataset.

For example, the 90th percentile indicates a value that is higher
than approximately 90% of the observations.

## 6. Interquartile Range (IQR)

### Definition

Interquartile Range (IQR) measures the spread of the middle 50%
of the data.

It is the difference between Q3 and Q1.

### Formula

IQR = Q3 - Q1
### Numerical Example

Consider:

10, 20, 30, 40, 50, 60, 70

Using percentile calculation:

Q1 = 25

Q3 = 55

Therefore:

IQR = 55 - 25

IQR = 30

```python
import numpy as np

data = [10, 20, 30, 40, 50, 60, 70]

q1 = np.percentile(data, 25)
q3 = np.percentile(data, 75)

iqr = q3 - q1

print("Q1:", q1)
print("Q3:", q3)
print("IQR:", iqr)
```

### Business Example

A company can use IQR to understand the spread of employee salaries
without being strongly affected by extremely high or low salaries.

### AI/ML Example

IQR is commonly used to detect outliers in machine learning datasets.

Outlier boundaries can be calculated as:

Lower Bound = Q1 - 1.5 × IQR

Upper Bound = Q3 + 1.5 × IQR

Values outside these boundaries are commonly considered potential
outliers.

### Interpretation

IQR represents the spread of the middle 50% of the data.

A larger IQR means the middle values are more spread out.

A smaller IQR means the middle values are closer together.

## 7. Coefficient of Variation

### Definition

Coefficient of Variation (CV) measures the relative amount of
variation compared with the mean.

It is especially useful when comparing datasets with different
units or different means.

### Formula

CV = (Standard Deviation / Mean) × 100

The result is expressed as a percentage.

### Numerical Example

Suppose:

Mean = 50

Standard Deviation = 5

CV = (5 / 50) × 100

CV = 10%

```python
mean = 50
standard_deviation = 5

cv = (standard_deviation / mean) * 100

print("Mean:", mean)
print("Standard Deviation:", standard_deviation)
print("Coefficient of Variation:", cv, "%")
```

### Business Example

Suppose two products have different average sales and different
levels of variation.

Coefficient of Variation can help compare their relative
variability.

### AI/ML Example

CV can be used during exploratory data analysis to compare the
relative variability of numerical features.

### Interpretation

A lower CV indicates less variation relative to the mean.

A higher CV indicates more variation relative to the mean.

CV is useful when comparing datasets with different scales.
