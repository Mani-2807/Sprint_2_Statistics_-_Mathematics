# HYPOTHESIS TESTING

# 1. Null Hypothesis

## Definition

The Null Hypothesis, represented by H₀, is a statement that there is no significant difference, effect, or relationship in the population.

It is the assumption that we test using statistical evidence.

### Example

Suppose a company introduces a new advertisement.

H₀: The new advertisement does not increase sales.

The hypothesis test will determine whether there is enough evidence to reject this assumption

## Numerical Example

Suppose the average monthly sales before a new advertisement were ₹50,000.

We want to check whether the new advertisement changed the average sales.

Null Hypothesis:

H₀: Mean sales = ₹50,000

This means we assume that the new advertisement has not changed the average sales.

```python
import numpy as np

sales = np.array([50000, 51000, 49000, 50500, 50000])

sample_mean = np.mean(sales)

print("Sample Mean:", sample_mean)
```

## Business Example

A company may want to test whether a new marketing campaign has increased average sales.

The null hypothesis can be:

H₀: The new marketing campaign does not increase average sales.

## AI/ML Example

In machine learning, the null hypothesis can be used when comparing two models or evaluating whether a change in a feature or process has a statistically significant effect.

For example:

H₀: The new model does not perform better than the existing model.

## Interpretation

The null hypothesis represents the default assumption.

Statistical testing uses sample data to determine whether there is enough evidence to reject H₀.

# 2. Alternative Hypothesis

## Definition

The Alternative Hypothesis, represented by H₁ or Hₐ, is a statement that there is a significant difference, effect, or relationship.

It is generally the opposite of the null hypothesis.

### Example

If:

H₀: The new advertisement does not increase sales.

Then:

H₁: The new advertisement increases sales.

## Numerical Example

Suppose the current average sales are ₹50,000.

We want to check whether a new marketing campaign increases sales.

H₀: Mean sales = ₹50,000

H₁: Mean sales > ₹50,000

The alternative hypothesis represents the expected increase.

```python
sample_mean = 55000
current_mean = 50000

if sample_mean > current_mean:
    print("Sample mean is higher than the current mean.")
else:
    print("Sample mean is not higher than the current mean.")
```

## Business Example

A company may want to know whether a new website design increases customer conversions.

H₀: The new website design does not increase conversions.

H₁: The new website design increases conversions.

## AI/ML Example

Suppose a new machine learning model is introduced.

H₀: The new model does not improve prediction performance.

H₁: The new model improves prediction performance.

## Interpretation

The alternative hypothesis represents the effect or difference that we are trying to find evidence for.

If there is enough statistical evidence, we reject the null hypothesis in favor of the alternative hypothesis.

# 3. p-value

## Definition

The p-value measures how likely it is to observe the obtained results, or more extreme results, if the null hypothesis is true.

A smaller p-value provides stronger evidence against the null hypothesis.

### Common Rule

If:

p-value < 0.05

→ Reject the Null Hypothesis.

If:

p-value ≥ 0.05

→ Fail to reject the Null Hypothesis.

The value 0.05 is a commonly used significance level.

## Numerical Example

Suppose a hypothesis test gives:

p-value = 0.03

Significance level = 0.05

Since:

0.03 < 0.05

We reject the null hypothesis.

This means there is statistically significant evidence against H₀.

```python
p_value = 0.03
alpha = 0.05

if p_value < alpha:
    print("Reject the Null Hypothesis")
else:
    print("Fail to Reject the Null Hypothesis")
```

## Business Example

A company tests whether a new advertisement increases sales.

If the p-value is 0.02, which is less than 0.05, the company has statistical evidence that the new advertisement may have an effect on sales.

## AI/ML Example

A p-value can be used when statistically comparing models or analyzing whether a feature has a significant relationship with a target.

A small p-value may provide evidence that the observed effect is statistically significant.

## Interpretation

A small p-value means the observed result would be relatively unlikely if the null hypothesis were true.

Important:

A p-value does not tell us the probability that the null hypothesis is true.

# 4. Confidence Interval

## Definition

A confidence interval is a range of values that is used to estimate an unknown population parameter.

A common confidence level is 95%.

A 95% confidence interval gives a range of plausible values for the population parameter based on the sample.

## Formula

For a population mean when the standard deviation is known:

CI = x̄ ± Z × (σ / √n)

Where:

- x̄ = Sample mean
- Z = Critical value
- σ = Population standard deviation
- n = Sample 

## Numerical Example

Suppose:

Sample Mean = 100

Standard Deviation = 10

Sample Size = 100

For a 95% confidence interval:

Z = 1.96

Margin of Error:

1.96 × (10 / √100)

= 1.96

Confidence Interval:

100 ± 1.96

Lower Limit = 98.04

Upper Limit = 101.96

```python
import math

sample_mean = 100
std = 10
n = 100
z = 1.96

margin_error = z * (std / math.sqrt(n))

lower = sample_mean - margin_error
upper = sample_mean + margin_error

print("Confidence Interval:", (lower, upper))
```

## Business Example

A company can estimate the average amount customers spend.

For example:

Estimated average spending = ₹1,000

95% Confidence Interval = ₹950 to ₹1,050

This provides a range of plausible values for the population average

## AI/ML Example

Confidence intervals can be used to understand the uncertainty around statistical estimates and model performance metrics.

For example, we can estimate a confidence interval around the accuracy of a model measured on a sample of data

## Interpretation

A confidence interval provides a range of plausible values for a population parameter.

A narrower interval generally indicates a more precise estimate.

# 5. Significance Level

## Definition

The significance level, represented by α (alpha), is the threshold used to decide whether a result is statistically significant.

A commonly used significance level is:

α = 0.05

This means we accept a 5% risk of rejecting the null hypothesis when it is actually true.

### Decision Rule

If:

p-value < α

→ Reject H₀

If:

p-value ≥ α

→ Fail to reject H₀

## Numerical Example

Suppose:

Significance Level = 0.05

p-value = 0.03

Since:

0.03 < 0.05

We reject the null hypothesis.

```python
alpha = 0.05
p_value = 0.03

if p_value < alpha:
    print("Statistically Significant")
else:
    print("Not Statistically Significant")
```

## Business Example

A company tests whether a new pricing strategy changes customer purchases.

Using α = 0.05, the company can decide whether the observed change is statistically significant.

## AI/ML Example

The significance level can be used when testing whether a feature or model improvement has a statistically significant effect.

## Interpretation

The significance level is the threshold used to make a statistical decision.

A smaller alpha value means stronger evidence is required to reject the null hypothesis.

# 6. Type I Error

## Definition

A Type I Error occurs when we reject the null hypothesis even though the null hypothesis is actually true.

It is also called a:

False Positive

### Example

H₀: A customer is not committing fraud.

If we reject H₀ and classify the customer as fraudulent when they are actually legitimate, this is a Type I Error.

## Relationship with Significance Level

The probability of making a Type I Error is represented by α.

For example:

α = 0.05

## Numerical Example

Suppose:

H₀: A new medicine has no effect.

The statistical test rejects H₀.

But in reality, the medicine has no effect.

This is a Type I Error.

```python
actual_null_true = True
reject_null = True

if actual_null_true and reject_null:
    print("Type I Error")
```

## Business Example

A fraud detection system flags a legitimate transaction as fraudulent.

This is a false positive and can be considered a Type I Error in the testing context.

## AI/ML Example

In a classification system, a Type I Error can occur when a model incorrectly identifies a normal observation as belonging to the positive class.

Example:

A normal transaction is incorrectly classified as fraudulent.

## Interpretation

Type I Error means:

We reject a true null hypothesis.

It is commonly associated with a false positive.

# 7. Type II Error

## Definition

A Type II Error occurs when we fail to reject the null hypothesis even though the null hypothesis is actually false.

It is also called a:

False Negative

### Example

H₀: A customer is not committing fraud.

If a fraudulent transaction is classified as normal, this represents a Type II Error.

### Relationship

The probability of Type II Error is represented by β (beta).

## Numerical Example

Suppose:

H₀: A new medicine has no effect.

In reality, the medicine does have an effect.

But the test fails to detect the effect and does not reject H₀.

This is a Type II Error.

```python
actual_null_false = True
fail_to_reject = True

if actual_null_false and fail_to_reject:
    print("Type II Error")
```

## Business Example

A fraud detection system fails to identify an actual fraudulent transaction.

This is a false negative and can represent a Type II Error.

## AI/ML Example

In a disease detection model, a Type II Error can occur when a person who actually has a disease is predicted as healthy.

This is an important example of a false negative

## Interpretation

Type II Error means:

We fail to reject a false null hypothesis.

It is commonly associated with a false negative.

# 8. One-Tailed Test

## Definition

A one-tailed test is a hypothesis test where the alternative hypothesis looks for an effect in only one direction.

The test can check whether a parameter is:

- Greater than a specific value, or
- Less than a specific value.

### Example

Suppose the current average sales are ₹50,000.

We want to test whether a new campaign increases sales.

H₀: μ ≤ ₹50,000

H₁: μ > ₹50,000

This is a right-tailed test.

## Numerical Example

Current mean = 50

Sample mean = 55

We want to determine whether the new process increases the mean.

H₀: μ ≤ 50

H₁: μ > 50

Because we are testing only for an increase, this is a one-tailed test.

    

```python
sample_mean = 55
current_mean = 50

if sample_mean > current_mean:
    print("Evidence of an increase")
else:
    print("No evidence of an increase")
```

## Business Example

A company wants to know whether a new marketing campaign increases sales.

Since the company is specifically interested in an increase, a one-tailed test can be used.

## AI/ML Example

Suppose we want to test whether a new model has higher accuracy than an existing model.

H₀: New model accuracy ≤ Existing model accuracy

H₁: New model accuracy > Existing model accuracy

This is a one-tailed test.

## Interpretation

A one-tailed test checks for an effect in one specific direction.

It is used when the research question clearly focuses on an increase or a decrease.

# 9. Two-Tailed Test

## Definition

A two-tailed test is a hypothesis test where the alternative hypothesis checks for a difference in either direction.

The test checks whether a parameter is:

- Greater than, or
- Less than

a specific value.

### Example

Current average sales = ₹50,000

We want to know whether a new campaign changes sales in either direction.

H₀: μ = ₹50,000

H₁: μ ≠ ₹50,000

This is a two-tailed test

## Numerical Example

Suppose:

Current Mean = 50

Sample Mean = 55

We want to know whether the mean has changed.

H₀: μ = 50

H₁: μ ≠ 50

Because we are checking for a difference in either direction, this is a two-tailed test.

```python
sample_mean = 55
current_mean = 50

if sample_mean != current_mean:
    print("There is a difference from the current mean")
else:
    print("No difference from the current mean")
```

## Business Example

A company wants to know whether a new pricing strategy changes customer spending.

The company is interested in both:

- An increase in spending
- A decrease in spending

Therefore, a two-tailed test can be used

## AI/ML Example

A two-tailed test can be used to determine whether a new machine learning model has a statistically different performance from an existing model.

The new model could perform either better or worse.

## Interpretation

A two-tailed test checks for a difference in either direction.

It is appropriate when we do not have a specific directional expectation.
