# OUTLIERS DETECTION

# 1. What are Outliers?

## Definition

An outlier is a data point that is significantly different from the other observations in a dataset.

For example:

Data = [10, 12, 11, 13, 12, 100]

Most values are between 10 and 13, but 100 is very different from the other values.

Therefore, 100 can be considered an outlier.

## Why are Outliers Important?

Outliers can:

- Affect the mean.
- Increase variance and standard deviation.
- Affect statistical analysis.
- Influence machine learning models.
- Sometimes represent errors or unusual real-world events.

## Important Note

An outlier is not always an error.

It may represent:
- A data entry mistake.
- An unusual event.
- A genuine extreme observation

## Numerical Example

Consider the following dataset:

[10, 12, 11, 13, 12, 100]

Most values are close to each other.

The value 100 is much larger than the other observations.

Therefore, 100 is a possible outlier.

```python
import numpy as np

data = [10, 12, 11, 13, 12, 100]

mean = np.mean(data)

print("Mean:", mean)
```

## Business Example

A company may have monthly sales values such as:

[₹50,000, ₹52,000, ₹48,000, ₹51,000, ₹5,00,000]

The value ₹5,00,000 is much higher than the other months.

It could be an outlier caused by a major festival sale or a data entry mistake.

The business should investigate the reason before removing it

## AI/ML Example

In a house price dataset, most houses may have prices between ₹30 lakh and ₹80 lakh, while one house may have a price of ₹5 crore.

This extreme value may affect the machine learning model.

Therefore, outlier detection is an important part of data preprocessing.

## Interpretation

An outlier is an observation that is unusually different from the rest of the data.

Before removing an outlier, we should understand why it occurred.

Some outliers are errors, while others represent important real-world events.

# 2. Types of Outliers

## Definition

Outliers can occur in different ways depending on the structure of the data.

### 1. Point Outlier

A single observation is significantly different from the other observations.

Example:

[10, 12, 11, 13, 100]

Here, 100 is a point outlier.

### 2. Contextual Outlier

A value may be unusual only in a particular context.

Example:

A temperature of 35°C may be normal in summer but unusual in winter.

### 3. Collective Outlier

A group of observations is unusual compared with the rest of the dataset.

Example:

A sudden series of unusual network requests may indicate suspicious activity.

## Numerical Example

Point Outlier:

[10, 11, 12, 13, 100]

Contextual Outlier:

A temperature of 40°C may be normal during summer but unusual during winter.

Collective Outlier:

A sudden group of unusual transactions in a short period may represent suspicious activity.

```python
data = [10, 11, 12, 13, 100]

print("Dataset:", data)
print("Potential outlier:", 100)
```

## Business Example

In banking:

- A single unusually large transaction can be a point outlier.
- A transaction that is unusual for a particular customer can be a contextual outlier.
- A group of unusual transactions within a short time can be a collective outlier

## AI/ML Example

Different types of outliers can be useful in:

- Fraud detection
- Network intrusion detection
- Anomaly detection
- Sensor monitoring

Machine learning systems can identify unusual patterns that may indicate abnormal behavior.


## Interpretation

Understanding the type of outlier helps us decide how to handle it.

A point outlier may be handled differently from a contextual or collective outlier.

# 3. Detecting Outliers

## Definition

Outlier detection is the process of identifying observations that are significantly different from the rest of the dataset.

Common methods include:

1. IQR Method
2. Z-Score Method
3. Box Plot
4. Scatter Plot

The choice of method depends on the data and the problem

## Numerical Example

Consider:

[10, 12, 11, 13, 12, 100]

The value 100 is much larger than the other values.

We can use statistical methods such as IQR or Z-score to determine whether 100 is an outlier.

```python
import numpy as np

data = np.array([10, 12, 11, 13, 12, 100])

Q1 = np.percentile(data, 25)
Q3 = np.percentile(data, 75)

IQR = Q3 - Q1

lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

outliers = data[(data < lower_limit) | (data > upper_limit)]

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Outliers:", outliers)
```

## Business Example

A company can use outlier detection to identify unusual:

- Sales transactions
- Customer spending
- Product prices
- Delivery times

This can help identify errors, unusual events, or suspicious activity

## AI/ML Example

Outlier detection is commonly used during data preprocessing.

Identifying extreme values can help improve the quality of data before training a machine learning model.

## Interpretation

Different statistical methods can be used to detect outliers.

The IQR method is particularly useful when the data contains extreme values or is not normally distributed.

# 4. IQR Method

## Definition

IQR stands for Interquartile Range.

It measures the spread of the middle 50% of the data.

## Formula

IQR = Q3 - Q1

Where:

- Q1 = First Quartile (25th percentile)
- Q3 = Third Quartile (75th percentile)

## Outlier Limits

Lower Bound = Q1 - 1.5 × IQR

Upper Bound = Q3 + 1.5 × IQR

Any value below the lower bound or above the upper bound is considered a potential outlier.

## Numerical Example

Data:

[10, 12, 11, 13, 12, 100]

Q1 = 11.25

Q3 = 12.75

IQR = Q3 - Q1

IQR = 12.75 - 11.25

IQR = 1.5

Upper Bound:

12.75 + (1.5 × 1.5) = 15

Since 100 > 15, 100 is an outlier.

```python
import numpy as np

data = np.array([10, 12, 11, 13, 12, 100])

Q1 = np.percentile(data, 25)
Q3 = np.percentile(data, 75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers = data[(data < lower_bound) | (data > upper_bound)]

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)
print("Outliers:", outliers)
```

## Business Example

A retail company can use the IQR method to identify unusually high or low customer spending.

For example, if most customers spend between ₹500 and ₹5,000 but one transaction is ₹1,00,000, the IQR method can help identify it as a potential outlier.

## AI/ML Example

The IQR method can be used during data preprocessing to detect extreme values in numerical features.

It is useful when the data is skewed or does not follow a normal distribution.

## Interpretation

The IQR method identifies values that are unusually far from the middle 50% of the data.

It is less affected by extreme values than methods based on the mean and standard deviation.

## Numerical Example

Suppose:

Mean = 50

Standard Deviation = 10

Data Point = 90

Z = (90 - 50) / 10

Z = 4

Since the Z-score is greater than +3, the value 90 can be considered a potential outlier.

```python
from scipy.stats import zscore
import numpy as np

data = np.array([10, 12, 11, 13, 12, 100])

z_scores = zscore(data)

print("Z-Scores:", z_scores)

outliers = data[np.abs(z_scores) > 3]

print("Outliers:", outliers)
```

### NOTE  
For this small dataset, the Z-score method does not identify 100 as an outlier using the ±3 rule, even though the IQR method does.

## Business Example

A company can use Z-scores to identify unusually high or low values in metrics such as:

- Daily sales
- Transaction amounts
- Delivery times
- Customer spending

## AI/ML Example

Z-scores can be used to identify unusual observations in numerical features, especially when the data is approximately normally distributed.

They can also be useful during data preprocessing and anomaly detection.

## Interpretation

A large positive Z-score means the value is far above the mean.

A large negative Z-score means the value is far below the mean.

The Z-score method works best when the data is approximately normally distributed.

# 6. Box Plot

## Definition

A box plot is a graphical method used to show the distribution of numerical data.

It displays:

- Minimum
- Q1
- Median
- Q3
- Maximum
- Potential outliers

Outliers are usually displayed as individual points outside the whiskers.

```python
import matplotlib.pyplot as plt

data = [10, 12, 11, 13, 12, 100]

plt.boxplot(data)

plt.title("Box Plot for Outlier Detection")
plt.ylabel("Values")

plt.show()
```

## Business Example

A company can use box plots to compare customer spending across different groups.

For example, a box plot can help identify customers whose spending is unusually high or low.

## AI/ML Example

Box plots are useful during Exploratory Data Analysis (EDA) to visually identify outliers in numerical features before training a machine learning model.

## Interpretation

A box plot provides a quick visual summary of the distribution of the data.

Points outside the whiskers are potential outliers and should be investigated.

# 7. Scatter Plot

## Definition

A scatter plot displays individual observations as points on a graph.

It is useful for understanding the relationship between two numerical variables.

Scatter plots can help identify:

- Positive relationships
- Negative relationships
- No clear relationship
- Clusters
- Potential outliers

## Numerical Example

Consider:

Study Hours = [1, 2, 3, 4, 5, 10]

Marks = [20, 40, 60, 80, 100, 30]

Most observations follow an increasing pattern.

The point (10, 30) is far away from the general pattern and may be an outlier.

```python
import matplotlib.pyplot as plt

study_hours = [1, 2, 3, 4, 5, 10]
marks = [20, 40, 60, 80, 100, 30]

plt.scatter(study_hours, marks)

plt.xlabel("Study Hours")
plt.ylabel("Marks")
plt.title("Scatter Plot for Outlier Detection")

plt.show()
```

## Business Example

A company can use a scatter plot to study the relationship between advertising expenditure and sales.

An unusual point may represent a month with unusually high advertising spending or unusually low sales.

## AI/ML Example

Scatter plots are useful during EDA to identify unusual observations and understand relationships between features.

For example, they can help identify unusual data points in a house price prediction dataset.

## Interpretation

A scatter plot makes it easy to visually identify observations that are far away from the general pattern.

Such observations can be investigated as potential outliers.
